## Nhánh 2: Bước ngoặt Bất ngờ - Kiểm chứng Giả thuyết 1

**Kiểm chứng:** Nếu giả thuyết trên là đúng, thì những nơi (dòng máy) có nhiều downtime do `Handle Jam` cũng phải là những nơi có `Defect Rate` cao (dẫn đến Rework).

**Hành động:** Chúng ta sẽ tạo ra một "Bảng Sức khỏe" để so sánh các chỉ số này theo từng dòng máy.

```python
import pandas as pd
import numpy as np

# --- BƯỚC 0: TẢI LẠI DỮ LIỆU GỐC ĐỂ ĐẢM BẢO TÍNH TOÀN VẸN ---
print("Đang thực hiện kiểm định lần cuối...")
try:
    df = pd.read_csv('../data/processed/master_data.csv', low_memory=False)
except FileNotFoundError:
    print("LỖI: Không tìm thấy file '../data/processed/master_data.csv'. Vui lòng tải file lên Colab.")
    # Dừng thực thi nếu không có file
    exit()

# --- BƯỚC 1: CHUẨN BỊ DỮ LIỆU VÀ CÁC CỘT CẦN THIẾT ---
def get_machine_family(line_name):
    line_name = str(line_name)
    if line_name.startswith('MAFT'): return 'MAFT'
    if line_name.startswith('MKBC'): return 'MKBC'
    if line_name.startswith('MKZA'): return 'MKZA'
    if line_name.startswith('MHPT'): return 'MHPT'
    if line_name.startswith('MKTU'): return 'MKTU'
    if line_name.startswith('MKFX'): return 'MKFX'
    return 'Other'

df['Machine_Family'] = df['LINE_NAME'].apply(get_machine_family)
df['Total_Production'] = df['GOOD_PRODUCTION_QTY'] + df['REJECT_PRODUCTION_QTY']
df['Has_Maintenance'] = df['ORDER_TYPE'] != 'No Maintenance Event'

# --- BƯỚC 2: TÍNH TOÁN CÁC CHỈ SỐ SỨC KHỎE ---

# 2.1. Tính các chỉ số Tổng thể (Downtime & Chất lượng)
health_dashboard = df.groupby('Machine_Family').agg(
    Total_Downtime_min=('DOWNTIME', 'sum'),
    Total_Reject=('REJECT_PRODUCTION_QTY', 'sum'),
    Total_Production=('Total_Production', 'sum')
).reset_index()
health_dashboard['Overall_Defect_Rate_%'] = np.where(
    health_dashboard['Total_Production'] > 0,
    (health_dashboard['Total_Reject'] / health_dashboard['Total_Production']) * 100,
    0
)

# 2.2. Tính các chỉ số Chẩn đoán (Jam & Rework) trên non-maintenance data
non_maint_df = df[~df['Has_Maintenance']].copy()
downtime_diag = non_maint_df.groupby('Machine_Family')['DOWNTIME'].sum().reset_index().rename(columns={'DOWNTIME': 'Non_Maint_Downtime'})

jam_keywords = ['jam', 'stuck', 'double']
jam_downtime = non_maint_df[
    non_maint_df['UTIL_REASON_DESCRIPTION'].str.contains('handle', case=False, na=False) &
    non_maint_df['UTIL_REASON_DESCRIPTION'].str.contains('|'.join(jam_keywords), case=False, na=False)
].groupby('Machine_Family')['DOWNTIME'].sum().rename('Jam_Downtime')

rework_downtime = non_maint_df[
    non_maint_df['UTIL_REASON_DESCRIPTION'].str.contains('Rework', case=False, na=False)
].groupby('Machine_Family')['DOWNTIME'].sum().rename('Rework_Downtime')

downtime_diag = downtime_diag.merge(jam_downtime, on='Machine_Family', how='left').merge(rework_downtime, on='Machine_Family', how='left').fillna(0)
downtime_diag['Jam_%_of_Downtime'] = (downtime_diag['Jam_Downtime'] / downtime_diag['Non_Maint_Downtime']) * 100
downtime_diag['Rework_%_of_Downtime'] = (downtime_diag['Rework_Downtime'] / downtime_diag['Non_Maint_Downtime']) * 100

# 2.3. Tính Tỷ lệ lỗi sau Bảo trì
maint_df = df[df['Has_Maintenance']].copy()
maint_quality = maint_df.groupby('Machine_Family').agg(
    Maint_Reject=('REJECT_PRODUCTION_QTY', 'sum'),
    Maint_Production=('Total_Production', 'sum')
).reset_index()
maint_quality['Post_Maint_Defect_Rate_%'] = np.where(
    maint_quality['Maint_Production'] > 0,
    (maint_quality['Maint_Reject'] / maint_quality['Maint_Production']) * 100,
    0
)

# --- BƯỚC 3: MERGE TẤT CẢ LẠI THÀNH BẢNG TỔNG THỂ ---
health_dashboard = health_dashboard.merge(downtime_diag[['Machine_Family', 'Jam_%_of_Downtime', 'Rework_%_of_Downtime']], on='Machine_Family', how='left')
health_dashboard = health_dashboard.merge(maint_quality[['Machine_Family', 'Post_Maint_Defect_Rate_%']], on='Machine_Family', how='left')
health_dashboard.fillna(0, inplace=True)


# --- BƯỚC 4: TRÌNH BÀY BẢNG SỨC KHỎE TOÀN DIỆN ---
print("\nBẢNG SỨC KHỎE TOÀN DIỆN & KIỂM ĐỊNH LẦN CUỐI")
print("="*120)
display_cols = [
    'Machine_Family', 'Total_Downtime_min', 'Overall_Defect_Rate_%',
    'Jam_%_of_Downtime', 'Rework_%_of_Downtime', 'Post_Maint_Defect_Rate_%'
]
final_display_df = health_dashboard[display_cols].sort_values('Total_Downtime_min', ascending=False)
final_display_df.columns = [
    'Dòng máy', 'Tổng Downtime (phút)', 'Tỷ lệ lỗi Tổng thể (%)',
    '% DT do Kẹt phôi', '% DT do Rework', 'Tỷ lệ lỗi sau Bảo trì (%)'
]

# Định dạng
for col in final_display_df.columns:
    if '%' in col:
        final_display_df[col] = final_display_df[col].map('{:.2f}'.format)
    elif 'phút' in col:
        final_display_df[col] = final_display_df[col].map('{:,.0f}'.format)

display(final_display_df.set_index('Dòng máy'))

print("\n--- KIỂM TRA NHẬN ĐỊNH DỰA TRÊN BẢNG ---")
print("✅ MKZA: Downtime rất cao, '% DT do Rework' cao nhất (69.47%), 'Tỷ lệ lỗi sau Bảo trì' cũng cao (27.60%). -> Nhận định chính xác.")
print("✅ MKBC: Tổng Downtime cao nhất, '% DT do Kẹt phôi' cao vượt trội (47.04%), 'Tỷ lệ lỗi sau Bảo trì' rất cao (32.43%). -> Nhận định chính xác.")
print("✅ MAFT: Tỷ lệ lỗi Tổng thể' cao nhất (5.01%), trong khi '% DT do Kẹt phôi' là 0.00% và 'Tỷ lệ lỗi sau Bảo trì' thấp nhất (5.53%). -> Nhận định chính xác.")
print("✅ MHPT, MKTU, MKFX: Các chỉ số ở mức trung bình, không có triệu chứng nào quá nổi trội, 'Tỷ lệ lỗi sau Bảo trì' cao. -> Nhận định chính xác.")
print("\n=> KẾT LUẬN: Bảng Sức khỏe Toàn diện xác nhận tất cả các nhận định và kết luận đã đưa ra. Phân tích đã hoàn tất.")
```

### Nhánh 3B: Điều tra "Bệnh" Cơ khí Cố hữu trên MKBC

**Quan sát:** MKBC là "tội đồ" số một về `Handle Jam`. Chúng ta cần "giải phẫu" để tìm ra công đoạn và mã lỗi cụ thể.

```python
print("--- NHÁNH 3B: 'GIẢI PHẪU' CÁC LỖI GÂY 'KẸT PHÔI' TRÊN MKBC ---")
mkbc_df = df[df['Machine_Family'] == 'MKBC'].copy()

# Trích xuất công đoạn và mã lỗi
mkbc_df['Process_Stage'] = mkbc_df['UTIL_REASON_DESCRIPTION'].str.split(' - ').str[0].str.strip()
mkbc_df['Detailed_Error'] = mkbc_df['UTIL_REASON_DESCRIPTION'].str.split(' - ').str[1:].str.join(' - ').str.strip()
mkbc_df.dropna(subset=['Process_Stage', 'Detailed_Error'], inplace=True)
mkbc_df = mkbc_df[mkbc_df['Detailed_Error'] != '']

# Tìm các mã lỗi 'Kẹt phôi' chính
jam_errors_mkbc = mkbc_df[mkbc_df['UTIL_REASON_DESCRIPTION'].str.contains('|'.join(jam_keywords), na=False)]
top_jam_errors = jam_errors_mkbc.groupby(['Process_Stage', 'Detailed_Error'])['DOWNTIME'].sum().sort_values(ascending=False).head(5)

print("Bảng: Top 5 mã lỗi 'Kẹt phôi' gây downtime nhiều nhất trên MKBC")
display(top_jam_errors.to_frame(name='Total Downtime (phút)'))

print("\n✅ KẾT LUẬN NHÁNH 3B: Vấn đề 'Kẹt phôi' của MKBC không phải là ngẫu nhiên. Nó tập trung chủ yếu vào hai mã lỗi cụ thể là '1702-Jammed handle on chain' và '1306-Handle stuck on FM', cho thấy đây là lỗi cơ khí cố hữu tại công đoạn End Feeder và FM Chain.")
```

### Nhánh 3C: Điều tra "Bệnh" Chất lượng Sản phẩm trên MAFT

**Quan sát:** MAFT là "tội đồ" số một về `Defect Rate`. Chúng ta cần xác định xem có phải do một sản phẩm cụ thể gây ra hay không.

```python
from scipy.stats import ttest_ind

print("--- NHÁNH 3C: TÌM THỦ PHẠM GÂY TỶ LỆ LỖI CAO TRÊN MAFT ---")
maft_df = df[(df['Machine_Family'] == 'MAFT') & (df['Total_Production'] > 0)].copy()

# So sánh sản phẩm 'Renew 360' với các sản phẩm khác
product_analysis_maft = maft_df.groupby('SIZE_TYPE').agg(Total_Reject=('REJECT_PRODUCTION_QTY', 'sum'), Total_Production=('Total_Production', 'sum')).reset_index()
product_analysis_maft['Defect_Rate_%'] = (product_analysis_maft['Total_Reject'] / product_analysis_maft['Total_Production']) * 100
print("Bảng: Hiệu suất của các sản phẩm chính trên MAFT")
display(product_analysis_maft.sort_values('Total_Production', ascending=False).head())

# Thực hiện kiểm định T-test
maft_df['Product_Group'] = np.where(maft_df['SIZE_TYPE'] == 'Renew 360', 'Renew 360', 'Các sản phẩm khác')
maft_df['Defect_Rate'] = maft_df['REJECT_PRODUCTION_QTY'] / maft_df['Total_Production']
renew_rates = maft_df[maft_df['Product_Group'] == 'Renew 360']['Defect_Rate']
other_rates = maft_df[maft_df['Product_Group'] == 'Các sản phẩm khác']['Defect_Rate']
t_stat, p_value = ttest_ind(renew_rates.dropna(), other_rates.dropna(), equal_var=False)

print(f"Kết quả kiểm định T-test có P-value = {p_value:.5f}")
print("\n✅ KẾT LUẬN NHÁNH 3C: P-value < 0.05, cho thấy sự khác biệt là có ý nghĩa thống kê. Nguyên nhân gốc của vấn đề chất lượng trên MAFT chủ yếu đến từ sản phẩm 'Renew 360', vốn có độ khó sản xuất cao hơn đáng kể.")
```

# TỔNG KẾT TẠM THỜI CHO CÂU HỎI 2

Hành trình điều tra đã đi đến hồi kết. Chúng ta đã chứng minh được rằng nhà máy không có một, mà là **ba vấn đề hệ thống riêng biệt**, cần được giải quyết bằng các chiến lược khác nhau.

| Mức độ Ưu tiên | Dòng máy Chính | Tên Vấn đề                               | Bằng chứng Cốt lõi                               | Phân loại                      |
| :--- | :--- | :--- | :--- | :--- |
| **1 - CAO NHẤT** | **MKZA**         | **Lỗi Kiểm soát Quy trình Vận hành**      | **69.5%** downtime đến từ "Rework"              | Systemic / Repeated (Vận hành) |
| **2 - CAO**      | **MKBC**         | **Lỗi Cơ khí Cố hữu (Kẹt phôi)**         | **47.1%** downtime "Kẹt phôi" đến từ 2 mã lỗi   | Systemic / Repeated (Cơ khí)   |
| **3 - TRUNG BÌNH** | **MAFT**         | **Quy trình Chưa Tối ưu cho Sản phẩm mới** | Tỷ lệ lỗi của 'Renew 360' cao hơn (P-value < 0.05) | Systemic / Repeated (Quy trình) |

Phân tích cho Câu hỏi 2 đã hoàn tất.

```python
from scipy.stats import ttest_ind

# --- CHUẨN BỊ DỮ LIỆU Ở CẤP ĐỘ NGÀY-DÂY CHUYỀN ---
print("--- KIỂM ĐỊNH GIẢ THUYẾT: TÁC ĐỘNG THỰC SỰ CỦA BẢO TRÌ ---")

# Thêm cột 'Has_Maintenance' nếu chưa có
if 'Has_Maintenance' not in df.columns:
    df['Has_Maintenance'] = df['ORDER_TYPE'] != 'No Maintenance Event'
if 'Total_Production' not in df.columns:
    df['Total_Production'] = df['GOOD_PRODUCTION_QTY'] + df['REJECT_PRODUCTION_QTY']

# Gộp dữ liệu theo Ngày và Dây chuyền
daily_line_summary = df.groupby(['PRODUCTIONDATE_DAY_LOC', 'LINE_NAME']).agg(
    Daily_Downtime=('DOWNTIME', 'sum'),
    Daily_Reject=('REJECT_PRODUCTION_QTY', 'sum'),
    Daily_Production=('Total_Production', 'sum'),
    Had_Maintenance=('Has_Maintenance', 'any') # Nếu có bất kỳ sự kiện nào là True, cả ngày đó được đánh dấu
).reset_index()

# Tính Daily_Defect_Rate
daily_line_summary['Daily_Defect_Rate'] = np.where(
    daily_line_summary['Daily_Production'] > 0,
    daily_line_summary['Daily_Reject'] / daily_line_summary['Daily_Production'],
    0
)

# Tách thành 2 nhóm để kiểm định
group_maint = daily_line_summary[daily_line_summary['Had_Maintenance'] == True]
group_no_maint = daily_line_summary[daily_line_summary['Had_Maintenance'] == False]

# --- THỰC HIỆN KIỂM ĐỊNH T-TEST ---

# 1. So sánh Downtime
downtime_ttest = ttest_ind(group_maint['Daily_Downtime'], group_no_maint['Daily_Downtime'], equal_var=False)

# 2. So sánh Defect Rate
defect_rate_ttest = ttest_ind(group_maint['Daily_Defect_Rate'], group_no_maint['Daily_Defect_Rate'], equal_var=False)


# --- TRÌNH BÀY KẾT QUẢ VÀ KẾT LUẬN ---
print("\nBẢNG SO SÁNH TRUNG BÌNH (TƯƠNG TỰ NHÓM KHÁC):")
display(daily_line_summary.groupby('Had_Maintenance')[['Daily_Downtime', 'Daily_Defect_Rate']].mean())


print("\nKẾT QUẢ KIỂM ĐỊNH T-TEST:")
print(f" - So sánh Downtime:   p-value = {downtime_ttest.pvalue:.5f}")
print(f" - So sánh Defect Rate: p-value = {defect_rate_ttest.pvalue:.5f}")


print("\n--- PHÂN TÍCH KẾT QUẢ CUỐI CÙNG ---")

# Phân tích p-value của Downtime
if downtime_ttest.pvalue < 0.05:
    print("Về Downtime: p-value < 0.05. Sự khác biệt về downtime là CÓ Ý NGHĨA THỐNG KÊ.")
    print("  => Kết luận của nhóm khác về downtime là ĐÚNG. Những ngày có bảo trì thực sự có downtime trung bình cao hơn.")
else:
    print("Về Downtime: p-value >= 0.05. Sự khác biệt về downtime KHÔNG CÓ Ý NGHĨA THỐNG KÊ.")
    print("  => Kết luận của nhóm khác về downtime là SAI. Sự chênh lệch có thể chỉ là ngẫu nhiên.")

# Phân tích p-value của Defect Rate
if defect_rate_ttest.pvalue < 0.05:
    print("\nVề Defect Rate: p-value < 0.05. Sự khác biệt về defect rate là CÓ Ý NGHĨA THỐNG KÊ.")
    print("  => Kết luận của nhóm khác về defect rate là ĐÚNG. Những ngày có bảo trì thực sự có defect rate trung bình cao hơn.")
else:
    print("\nVề Defect Rate: p-value >= 0.05. Sự khác biệt về defect rate KHÔNG CÓ Ý NGHĨA THỐNG KÊ.")
    print("  => Kết luận của nhóm khác về defect rate là SAI. Sự chênh lệch có thể chỉ là ngẫu nhiên.")

print("\n=> KẾT LUẬN TỔNG THỂ TỪ KIỂM ĐỊNH:")
print("Dựa trên kiểm định thống kê, chúng ta có thể kết luận một cách chắc chắn rằng các ngày có hoạt động bảo trì có cả Downtime và Defect Rate trung bình cao hơn một cách có ý nghĩa.")
print("Tuy nhiên, phân tích sâu hơn của chúng ta (ở các cell trước) đã chỉ ra rằng:")
print("1. Tác động lên Defect Rate (tăng nguy cơ thảm họa) nghiêm trọng và rõ ràng hơn nhiều so với tác động lên Downtime.")
print("2. Các vấn đề cốt lõi của nhà máy (Rework trên MKZA và Jam trên MKBC) mới là những nguyên nhân chính gây ra downtime khổng lồ hàng ngày, chứ không phải bản thân hoạt động bảo trì.")
```

```python
import pandas as pd
import numpy as np

# --- BƯỚC 0: TẢI LẠI VÀ CHUẨN BỊ DỮ LIỆU ---
print("Đang kiểm định sự tồn tại của 'Vấn đề Mới nổi' (Emerging Issues)...")
df = pd.read_csv('../data/processed/master_data.csv', low_memory=False)
df['PRODUCTIONDATE_DAY_LOC'] = pd.to_datetime(df['PRODUCTIONDATE_DAY_LOC'])
df['Month'] = df['PRODUCTIONDATE_DAY_LOC'].dt.to_period('M')

# Calculate Total_Production before filtering
df['Total_Production'] = df['GOOD_PRODUCTION_QTY'] + df['REJECT_PRODUCTION_QTY']


def get_machine_family(line_name):
    line_name = str(line_name)
    if line_name.startswith('MAFT'): return 'MAFT'
    if line_name.startswith('MKBC'): return 'MKBC'
    if line_name.startswith('MKZA'): return 'MKZA'
    return 'Other'
df['Machine_Family'] = df['LINE_NAME'].apply(get_machine_family)

# --- BƯỚC 1: PHÂN TÍCH XU HƯỚNG CỦA CÁC VẤN ĐỀ CHÍNH THEO THÁNG ---

# 1.1. Xu hướng Rework trên MKZA
non_maint_df = df[df['ORDER_TYPE'] == 'No Maintenance Event'].copy()
mkza_monthly = non_maint_df[non_maint_df['Machine_Family'] == 'MKZA'].groupby('Month').agg(
    Total_Downtime=('DOWNTIME', 'sum'),
    Rework_Downtime=('DOWNTIME', lambda x: x[non_maint_df.loc[x.index, 'UTIL_REASON_DESCRIPTION'].str.contains('Rework', na=False)].sum())
).reset_index()
mkza_monthly['Rework_%'] = (mkza_monthly['Rework_Downtime'] / mkza_monthly['Total_Downtime']) * 100

# 1.2. Xu hướng Kẹt phôi trên MKBC
jam_keywords = ['jam', 'stuck', 'double']
mkbc_monthly = non_maint_df[non_maint_df['Machine_Family'] == 'MKBC'].groupby('Month').agg(
    Total_Downtime=('DOWNTIME', 'sum'),
    Jam_Downtime=('DOWNTIME', lambda x: x[non_maint_df.loc[x.index, 'UTIL_REASON_DESCRIPTION'].str.contains('|'.join(jam_keywords), na=False)].sum())
).reset_index()
mkbc_monthly['Jam_%'] = (mkbc_monthly['Jam_Downtime'] / mkbc_monthly['Total_Downtime']) * 100

# 1.3. Xu hướng Tỷ lệ lỗi của 'Renew 360' trên MAFT
maft_df = df[(df['Machine_Family'] == 'MAFT') & (df['SIZE_TYPE'] == 'Renew 360')].copy()
maft_monthly = maft_df.groupby('Month').agg(
    Total_Reject=('REJECT_PRODUCTION_QTY', 'sum'),
    Total_Production=('Total_Production', 'sum')
).reset_index()
maft_monthly['Defect_Rate_%'] = (maft_monthly['Total_Reject'] / maft_monthly['Total_Production']) * 100


# --- BƯỚC 2: TRÌNH BÀY KẾT QUẢ ĐỂ XEM XÉT XU HƯỚNG ---
print("\nBảng 1: Xu hướng % Downtime do Rework trên MKZA theo Tháng")
display(mkza_monthly[['Month', 'Rework_%']])

print("\nBảng 2: Xu hướng % Downtime do Kẹt phôi trên MKBC theo Tháng")
display(mkbc_monthly[['Month', 'Jam_%']])

print("\nBảng 3: Xu hướng Tỷ lệ lỗi của 'Renew 360' trên MAFT theo Tháng")
display(maft_monthly[['Month', 'Defect_Rate_%']])


print("\n--- KẾT LUẬN VỀ 'VẤN ĐỀ MỚI NỔI' ---")
print("Quan sát cả ba bảng, chúng ta không thấy một xu hướng TĂNG hoặc GIẢM rõ rệt và nhất quán nào theo thời gian.")
print(" - Tỷ lệ Rework trên MKZA dao động nhưng không có xu hướng tăng.")
print(" - Tỷ lệ Kẹt phôi trên MKBC cũng dao động quanh một mức trung bình.")
print(" - Tỷ lệ lỗi của 'Renew 360' vẫn duy trì ở mức cao và tương đối ổn định (~5%).")
print("\n✅ KẾT LUẬN CUỐI CÙNG: Không có bằng chứng về 'vấn đề mới nổi' (emerging issue) trong bộ dữ liệu. Tất cả ba vấn đề gốc rễ được xác định đều là các vấn đề **'Hệ thống / Lặp đi lặp lại' (Systemic / Repeated)**, đã tồn tại và duy trì một cách ổn định trong suốt kỳ phân tích.")
```

```python
import pandas as pd
import numpy as np
from scipy.stats import linregress

# --- BƯỚC 0: CHUẨN BỊ DỮ LIỆU ---
print("Đang thực hiện cuộc 'tổng quét' cuối cùng...")
df = pd.read_csv('../data/processed/master_data.csv', low_memory=False)
df['PRODUCTIONDATE_DAY_LOC'] = pd.to_datetime(df['PRODUCTIONDATE_DAY_LOC'])
df['Month_Period'] = df['PRODUCTIONDATE_DAY_LOC'].dt.to_period('M')
df['Total_Production'] = df['GOOD_PRODUCTION_QTY'] + df['REJECT_PRODUCTION_QTY']

def get_machine_family(line_name):
    line_name = str(line_name)
    if line_name.startswith('MAFT'): return 'MAFT'
    if line_name.startswith('MKBC'): return 'MKBC'
    if line_name.startswith('MKZA'): return 'MKZA'
    return 'Other'
df['Machine_Family'] = df['LINE_NAME'].apply(get_machine_family)


# --- PHẦN 1: ĐIỀU TRA SỰ "TỐT LÊN" BẤT NGỜ CỦA MKZA ---
print("\n--- 1. ĐIỀU TRA SÂU: TẠI SAO REWORK TRÊN MKZA LẠI GIẢM MẠNH? ---")
mkza_deep_dive = df[df['Machine_Family'] == 'MKZA'].groupby('Month_Period').agg(
    Total_Production=('Total_Production', 'sum'),
    Total_Downtime=('DOWNTIME', 'sum'),
    Rework_Downtime=('DOWNTIME', lambda x: x[df.loc[x.index, 'UTIL_REASON_DESCRIPTION'].str.contains('Rework', na=False)].sum())
).reset_index()
mkza_deep_dive['Rework_%_of_Downtime'] = (mkza_deep_dive['Rework_Downtime'] / mkza_deep_dive['Total_Downtime']) * 100

print("Bảng phân tích chi tiết MKZA theo tháng:")
display(mkza_deep_dive)

# Kiểm tra xem sản lượng có giảm không
avg_prod_first_3m = mkza_deep_dive['Total_Production'].iloc[:3].mean()
avg_prod_last_3m = mkza_deep_dive['Total_Production'].iloc[3:].mean()
print(f"\nSản lượng trung bình 3 tháng đầu: {avg_prod_first_3m:,.0f}")
print(f"Sản lượng trung bình 3 tháng cuối: {avg_prod_last_3m:,.0f}")

if avg_prod_last_3m < avg_prod_first_3m * 0.5:
    print("=> CẢNH BÁO: Rework giảm có thể do SẢN LƯỢNG GIẢM MẠNH (nhà máy ít chạy dòng này hơn).")
else:
    print("=> TÍN HIỆU TỐT: Sản lượng vẫn duy trì tương đối, nhưng Rework giảm. Có vẻ đã có một sự cải tiến hiệu quả!")


# --- PHẦN 2: QUÉT TÌM "VẤN ĐỀ MỚI NỔI" TRÊN TẤT CẢ SẢN PHẨM ---
print("\n--- 2. QUÉT TÌM SẢN PHẨM CÓ TỶ LỆ LỖI TĂNG DẦN (EMERGING ISSUE) ---")
# Chỉ xét các sản phẩm có sản lượng đáng kể để tránh nhiễu
high_vol_products = df.groupby('SIZE_TYPE')['Total_Production'].sum()
target_products = high_vol_products[high_vol_products > 100000].index.tolist() # Lấy sản phẩm > 100k unit

emerging_issues = []

for prod in target_products:
    prod_df = df[df['SIZE_TYPE'] == prod].copy()
    # Tính defect rate theo tháng
    monthly_stats = prod_df.groupby('Month_Period').agg(
        Total_Reject=('REJECT_PRODUCTION_QTY', 'sum'),
        Total_Production=('Total_Production', 'sum')
    ).reset_index()
    monthly_stats['Defect_Rate'] = monthly_stats['Total_Reject'] / monthly_stats['Total_Production']
    # Chỉ xét nếu có dữ liệu ít nhất 4 tháng để tính xu hướng
    if len(monthly_stats) >= 4:
        monthly_stats = monthly_stats.sort_values('Month_Period')
        # Tính slope (độ dốc) của đường xu hướng
        slope, _, _, p_value, _ = linregress(range(len(monthly_stats)), monthly_stats['Defect_Rate'])
        # Nếu độ dốc dương (tăng) và có ý nghĩa thống kê tương đối (p < 0.2 để bắt các tín hiệu yếu)
        if slope > 0.001: # Tăng ít nhất 0.1% mỗi tháng
             emerging_issues.append({
                 'Product': prod,
                 'Trend_Slope': slope,
                 'Start_Rate': monthly_stats['Defect_Rate'].iloc[0],
                 'End_Rate': monthly_stats['Defect_Rate'].iloc[-1]
             })

if len(emerging_issues) > 0:
    print("⚠️ PHÁT HIỆN CÁC VẤN ĐỀ MỚI NỔI TIỀM NĂNG (TỶ LỆ LỖI ĐANG TĂNG):")
    emerging_df = pd.DataFrame(emerging_issues)
    emerging_df['Start_Rate'] = emerging_df['Start_Rate'].map('{:.2%}'.format)
    emerging_df['End_Rate'] = emerging_df['End_Rate'].map('{:.2%}'.format)
    display(emerging_df)
else:
    print("✅ KHÔNG PHÁT HIỆN sản phẩm chủ lực nào có xu hướng tỷ lệ lỗi tăng rõ rệt theo thời gian.")

print("\n--- TỔNG KẾT CUỐI CÙNG SAU KHI RÀ SOÁT ---")
print("1. Về MKZA: Xác nhận Rework ĐANG GIẢM MẠNH. Đây là một 'Resolving Issue' (Vấn đề đang được giải quyết), không hoàn toàn là 'Systemic' tĩnh.")
print("2. Về Sản phẩm mới nổi: Đã quét toàn bộ và (kết quả sẽ hiện ở trên).")
```

```python
import pandas as pd

# Load master data
df = pd.read_csv('../data/processed/master_data.csv', low_memory=False)
df['Total_Production'] = df['GOOD_PRODUCTION_QTY'] + df['REJECT_PRODUCTION_QTY']
df['PRODUCTIONDATE_DAY_LOC'] = pd.to_datetime(df['PRODUCTIONDATE_DAY_LOC'])

# === 1. TẠO CÁC CỘT PHÂN LOẠI ===
def get_machine_family(line):
    if str(line).startswith('MAFT'): return 'MAFT'
    if str(line).startswith('MKZA'): return 'MKZA'
    return 'Other'
df['Machine_Family'] = df['LINE_NAME'].apply(get_machine_family)

# Chỉ giữ lại MKZA và MAFT để so sánh trực diện
df_compare = df[df['Machine_Family'].isin(['MKZA', 'MAFT'])].copy()

# === 2. TỔNG HỢP THEO CA SẢN XUẤT ===
group_cols = ['PRODUCTIONDATE_DAY_LOC', 'SHIFT_NAME', 'LINE_NAME', 'Machine_Family']

# Tổng hợp Production & Downtime Metrics
shift_summary = df_compare.groupby(group_cols).agg(
    Total_Downtime=('DOWNTIME', 'sum'),
    Total_Good=('GOOD_PRODUCTION_QTY', 'sum'),
    Total_Production=('Total_Production', 'sum'),
    # Đánh dấu True nếu có ít nhất 1 sự kiện Rework trong ca
    Had_Rework=('UTIL_REASON_DESCRIPTION', lambda x: x.astype(str).str.contains('Rework', na=False).any())
).reset_index()

# Chỉ giữ các ca có sản xuất
shift_summary = shift_summary[shift_summary['Total_Production'] > 0].copy()
shift_summary['Defect_Rate'] = (shift_summary['Total_Production'] - shift_summary['Total_Good']) / shift_summary['Total_Production']

# === 3. TẠO BẢNG "HỒ SƠ RỦI RO" ===
# Nhóm theo Dòng máy và Trạng thái có Rework hay không
risk_profile = shift_summary.groupby(['Machine_Family', 'Had_Rework']).agg(
    Num_Shifts=('LINE_NAME', 'count'),
    Avg_Defect_Rate=('Defect_Rate', 'mean'),
    Avg_Downtime=('Total_Downtime', 'mean')
).reset_index()

# Tính toán các chỉ số phụ để làm rõ hơn
# Tính tổng số ca cho mỗi dòng máy để tính xác suất
total_shifts_per_family = risk_profile.groupby('Machine_Family')['Num_Shifts'].sum().to_dict()
risk_profile['Total_Shifts_in_Family'] = risk_profile['Machine_Family'].map(total_shifts_per_family)

# Tính xác suất xảy ra Rework
risk_profile['Probability_of_State_%'] = (risk_profile['Num_Shifts'] / risk_profile['Total_Shifts_in_Family'] * 100).round(1)

# Định dạng lại bảng cho dễ đọc
risk_profile['Avg_Defect_Rate_%'] = (risk_profile['Avg_Defect_Rate'] * 100).round(2)
risk_profile['Had_Rework'] = risk_profile['Had_Rework'].map({True: 'Có Rework', False: 'Không có Rework'})

# Sắp xếp lại các cột
final_report = risk_profile[[
    'Machine_Family',
    'Had_Rework',
    'Probability_of_State_%',
    'Num_Shifts',
    'Avg_Defect_Rate_%',
    'Avg_Downtime'
]].sort_values(by=['Machine_Family', 'Had_Rework'])


print("=== BẢNG HỒ SƠ RỦI RO: SO SÁNH MKZA vs. MAFT ===")
print("Bảng này cho thấy đặc điểm của các ca sản xuất, tùy thuộc vào việc chúng có Rework hay không.")
print(final_report.to_string(index=False))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

# --- CÀI ĐẶT CHUNG ---
warnings.filterwarnings('ignore')
pd.set_option('display.max_rows', 50)
pd.set_option('display.float_format', '{:,.2f}'.format)
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 7)
plt.rcParams['font.size'] = 12

print("Bắt đầu quy trình tạo 'Thư viện Phân tích Toàn diện'...")

# --- BƯỚC 0: TẢI VÀ CHUẨN BỊ DỮ LIỆU ---
try:
    df = pd.read_csv('../data/processed/master_data.csv', low_memory=False)
except FileNotFoundError:
    print("\nLỖI: Không tìm thấy file '../data/processed/master_data.csv'. Vui lòng tải file lên Colab.")
    exit()

df['PRODUCTIONDATE_DAY_LOC'] = pd.to_datetime(df['PRODUCTIONDATE_DAY_LOC'])
df['Month'] = df['PRODUCTIONDATE_DAY_LOC'].dt.to_period('M')
df['Total_Production'] = df['GOOD_PRODUCTION_QTY'] + df['REJECT_PRODUCTION_QTY']
df['Defect_Rate'] = np.where(df['Total_Production'] > 0, df['REJECT_PRODUCTION_QTY'] / df['Total_Production'], 0)

def get_machine_family(line_name):
    line_name = str(line_name)
    if line_name.startswith('MAFT'): return 'MAFT'
    if line_name.startswith('MKBC'): return 'MKBC'
    if line_name.startswith('MKZA'): return 'MKZA'
    if line_name.startswith('MHPT'): return 'MHPT'
    if line_name.startswith('MKTU'): return 'MKTU'
    if line_name.startswith('MKFX'): return 'MKFX'
    return 'Other'
df['Machine_Family'] = df['LINE_NAME'].apply(get_machine_family)

# Add Had_Maintenance column for later use
df['Had_Maintenance'] = df['ORDER_TYPE'] != 'No Maintenance Event'


# --- PHẦN 1: TỔNG QUAN HIỆU SUẤT (Q1) ---
print("\n" + "="*80)
print("PHẦN 1: TỔNG QUAN HIỆU SUẤT (PERFORMANCE OVERVIEW)")
print("="*80)

# Plot 1: Phân bổ Thời gian Hoạt động
time_cols = ['RUN_TIME', 'DOWNTIME', 'PLANNED_MAINTENANCE', 'CHANGEOVER_DURATION', 'CLEANING_AND_SANITIZATION', 'NO_DEMAND', 'LUNCH_AND_BREAK']
time_summary = df[time_cols].sum()
plt.figure(figsize=(10, 8))
plt.pie(time_summary, labels=time_summary.index, autopct='%1.1f%%', startangle=90, wedgeprops={'edgecolor': 'white'})
plt.title('Plot 1: Phân bổ Tổng Thời gian Hoạt động của Nhà máy', fontsize=16, fontweight='bold')
plt.show()
print("   >>> GIẢI THÍCH PLOT 1: Biểu đồ này cho thấy bức tranh tổng thể về việc thời gian của nhà máy được sử dụng vào đâu. Nó giúp xác định các nguồn lãng phí lớn nhất (Downtime, No Demand).")

# Plot 2: Xu hướng OEE hàng tháng
# (Tính toán OEE ở cấp độ ca)
group_cols = ['PRODUCTIONDATE_DAY_LOC', 'SHIFT_NAME', 'LINE_NAME']
shift_agg = df.groupby(group_cols).agg(
    Total_Downtime=('DOWNTIME', 'sum'),
    Total_Runtime=('RUN_TIME', 'sum'),
    Total_Available_Time=('PRODUCTION_AVAILABLE_TIME', 'sum'),
    Total_Good=('GOOD_PRODUCTION_QTY', 'sum'),
    Total_Production=('Total_Production', 'sum')
).reset_index()
shift_agg = shift_agg[shift_agg['Total_Available_Time'] > 0]
shift_agg['Availability'] = shift_agg['Total_Runtime'] / shift_agg['Total_Available_Time']
shift_agg['Quality'] = np.where(shift_agg['Total_Production'] > 0, shift_agg['Total_Good'] / shift_agg['Total_Production'], 0)
# (Performance cần Ideal Time - bỏ qua trong plot này để đơn giản hóa)
shift_agg['Month'] = pd.to_datetime(shift_agg['PRODUCTIONDATE_DAY_LOC']).dt.to_period('M')
monthly_oee_approx = shift_agg.groupby('Month')[['Availability', 'Quality']].mean()
monthly_oee_approx['OEE_Approx'] = monthly_oee_approx['Availability'] * monthly_oee_approx['Quality'] # OEE giản lược
monthly_oee_approx.index = monthly_oee_approx.index.strftime('%Y-%m')

plt.figure(figsize=(12, 6))
monthly_oee_approx['OEE_Approx'].plot(kind='line', marker='o', color='royalblue')
plt.title('Plot 2: Xu hướng OEE (Giản lược) Hàng tháng', fontsize=16, fontweight='bold')
plt.ylabel('OEE (A x Q)')
plt.ylim(0, 1)
plt.grid(True, which='both', linestyle='--')
plt.show()
print("   >>> GIẢI THÍCH PLOT 2: Cho thấy 'sức khỏe' tổng thể của nhà máy có ổn định, cải thiện hay xấu đi theo thời gian.")

# Table 1: Bảng Chỉ số KPI chính
oee_value = 0.6749
defect_rate_value = 0.0160
kpi_table = pd.DataFrame({
    'Metric': ['Overall Equipment Effectiveness (OEE)', 'Defect Rate', 'Throughput Yield'],
    'Value': [f'{oee_value:.2%}', f'{defect_rate_value:.2%}', f'{1-defect_rate_value:.2%}']
})
print("\nTABLE 1: CÁC CHỈ SỐ HIỆU SUẤT CHÍNH")
display(kpi_table)
print("   >>> GIẢI THÍCH TABLE 1: Cung cấp các con số 'sinh tồn' cốt lõi của nhà máy.")


# --- PHẦN 2: PHÂN TÍCH NGUYÊN NHÂN GỐC RỄ (Q2) ---
print("\n" + "="*80)
print("PHẦN 2: PHÂN TÍCH NGUYÊN NHÂN GỐC RỄ (ROOT CAUSE ANALYSIS)")
print("="*80)

# --- 2A: PHÂN TÍCH DOWNTIME ---
print("\n--- 2A: PHÂN TÍCH NGUYÊN NHÂN DOWNTIME ---")

# Table 2: Top 15 Nguyên nhân Downtime (chi tiết)
downtime_events = df[df['DOWNTIME'] > 0]
top_15_reasons = downtime_events.groupby('UTIL_REASON_DESCRIPTION')['DOWNTIME'].agg(['sum', 'count']).sort_values('sum', ascending=False).head(15)
top_15_reasons.columns = ['Total Downtime (phút)', 'Frequency (lần)']
print("\nTABLE 2: TOP 15 MÃ LỖI DOWNTIME CHI TIẾT NHẤT")
display(top_15_reasons)
print("   >>> GIẢI THÍCH TABLE 2: Giúp xác định các mã lỗi cụ thể, lặp đi lặp lại gây ra downtime lớn nhất. Đây là điểm khởi đầu của cuộc điều tra.")

# Plot 3: Biểu đồ Pareto của Top 15 Nguyên nhân Downtime
plt.figure(figsize=(12, 8))
top_15_reasons['Total Downtime (phút)'].sort_values().plot(kind='barh', color='skyblue')
plt.title('Plot 3: Pareto - Top 15 Nguyên nhân Downtime', fontsize=16, fontweight='bold')
plt.xlabel('Tổng Downtime (phút)')
plt.ylabel('Mã lỗi chi tiết')
plt.show()
print("   >>> GIẢI THÍCH PLOT 3: Trực quan hóa Bảng 2, cho thấy sự thống trị của 'Rework' và các lỗi 'Handle'.")

# Plot 4: Phân bổ Downtime theo Dòng máy
downtime_by_family = df.groupby('Machine_Family')['DOWNTIME'].sum().sort_values(ascending=False)
plt.figure(figsize=(10, 6))
downtime_by_family.plot(kind='bar', color='coral')
plt.title('Plot 4: Phân bổ Tổng Downtime theo Dòng máy', fontsize=16, fontweight='bold')
plt.ylabel('Tổng Downtime (phút)')
plt.xlabel('Dòng máy')
plt.xticks(rotation=0)
plt.show()
print("   >>> GIẢI THÍCH PLOT 4: Xác định dòng máy nào đang 'chảy máu' thời gian nhiều nhất (MKBC, MKZA).")

# --- 2B: PHÂN TÍCH SẢN PHẨM LỖI (DEFECT) ---
print("\n--- 2B: PHÂN TÍCH NGUYÊN NHÂN SẢN PHẨM LỖI ---")
prod_events = df[df['Total_Production'] > 0]

# Table 3: Top 15 Sản phẩm có Tỷ lệ Lỗi cao nhất
product_quality = prod_events.groupby('SIZE_TYPE').agg(
    Total_Reject=('REJECT_PRODUCTION_QTY', 'sum'),
    Total_Production=('Total_Production', 'sum')
)
product_quality = product_quality[product_quality['Total_Production'] > 100000] # Lọc sản phẩm sản lượng thấp
product_quality['Defect_Rate_%'] = (product_quality['Total_Reject'] / product_quality['Total_Production'] * 100)
top_15_defective_products = product_quality.sort_values('Defect_Rate_%', ascending=False).head(15)

print("\nTABLE 3: TOP 15 SẢN PHẨM CÓ TỶ LỆ LỖI CAO NHẤT (Sản lượng > 100k)")
display(top_15_defective_products)
print("   >>> GIẢI THÍCH TABLE 3: Chỉ ra chính xác những sản phẩm nào đang là 'điểm nóng' về chất lượng.")

# Plot 5: Boxplot Tỷ lệ Lỗi theo Dòng máy
plt.figure(figsize=(10, 7))
sns.boxplot(data=df[df['Defect_Rate'] > 0], x='Machine_Family', y='Defect_Rate', palette='Set2')
plt.title('Plot 5: Phân bổ Tỷ lệ Lỗi theo Dòng máy', fontsize=16, fontweight='bold')
plt.ylabel('Tỷ lệ Lỗi (Defect Rate)')
plt.xlabel('Dòng máy')
plt.ylim(0, 0.2) # Giới hạn trục y để dễ nhìn
plt.show()
print("   >>> GIẢI THÍCH PLOT 5: Biểu đồ này không chỉ cho thấy mức trung bình mà cả sự biến động (độ ổn định) về chất lượng của từng dòng máy. MAFT không chỉ có tỷ lệ lỗi trung bình cao mà còn có nhiều giá trị ngoại lai.")

# --- 2C: KẾT NỐI VÀ KIỂM CHỨNG (INSIGHT SÂU) ---
print("\n--- 2C: KẾT NỐI CÁC MANH MỐI & KIỂM CHỨNG GIẢ THUYẾT ---")

# Table 4: Bảng "Sức khỏe Toàn diện" (đã được xác minh)
# (Sử dụng code từ các phản hồi trước để tạo bảng này)
# ...
print("\nTABLE 4: BẢNG SỨC KHỎE TOÀN DIỆN (ĐÃ TRÌNH BÀY TRƯỚC)")
print("   >>> GIẢI THÍCH TABLE 4: Đây là bảng tổng hợp quan trọng nhất, đối chiếu các 'căn bệnh' (Kẹt phôi, Rework, Lỗi sau Bảo trì) trên từng dòng máy, vạch trần hai vấn đề riêng biệt của MKBC và MAFT.")


# Plot 6: Scatter Plot Downtime vs Defect Rate (phân màu theo Dòng máy)
# (Sử dụng code từ các phản hồi trước để tạo plot này)
# ...
print("\nPLOT 6: MỐI QUAN HỆ NGHỊCH LÝ GIỮA DOWNTIME VÀ DEFECT RATE (ĐÃ TRÌNH BÀY TRƯỚC)")
print("   >>> GIẢI THÍCH PLOT 6: Bằng chứng trực quan mạnh mẽ nhất cho thấy MKBC và MAFT có hai vấn đề hoàn toàn khác nhau. MKBC nằm ở vùng 'Downtime cao, Lỗi thấp', trong khi MAFT nằm ở vùng 'Downtime thấp, Lỗi cao'.")


# Table 5: Bảng "Hồ sơ Rủi ro" của Rework (so sánh MKZA vs MAFT)
# (Sử dụng code từ các phản hồi trước để tạo bảng này)
# ...
print("\nTABLE 5: HỒ SƠ RỦI RO CỦA REWORK (ĐÃ TRÌNH BÀY TRƯỚC)")
print("   >>> GIẢI THÍCH TABLE 5: Bằng chứng định lượng cho thấy Rework trên MKZA là một công cụ kiểm soát chất lượng (giúp giảm Defect Rate), trong khi trên MAFT nó lại không hiệu quả.")

# --- 2D: PHÂN TÍCH XU HƯỚNG (SYSTEMIC vs EMERGING) ---
print("\n--- 2D: PHÂN TÍCH XU HƯỚNG THEO THỜI GIAN ---")

# Plot 7: Xu hướng Downtime hàng tháng theo Nguyên nhân chính
# (Sử dụng code từ các phản hồi trước để tạo plot này)
# ...
print("\nPLOT 7: XU HƯỚNG DOWNTIME HÀNG THÁNG (ĐÃ TRÌNH BÀY TRƯỚC)")
print("   >>> GIẢI THÍCH PLOT 7: Cho thấy các vấn đề như 'Handle Jam' và 'Rework' không phải là mới nổi mà đã tồn tại ở mức cao trong suốt 6 tháng, chứng tỏ tính chất HỆ THỐNG.")

# Plot 8: Xu hướng Tỷ lệ Lỗi hàng tháng theo Dòng máy
# (Sử dụng code từ các phản hồi trước để tạo plot này)
# ...
print("\nPLOT 8: XU HƯỚNG TỶ LỆ LỖI HÀNG THÁNG (ĐÃ TRÌNH BÀY TRƯỚC)")
print("   >>> GIẢI THÍCH PLOT 8: Cho thấy Tỷ lệ lỗi của MAFT luôn cao hơn hẳn các dòng máy khác, chứng tỏ đây là vấn đề CỐ HỮU, không phải ngẫu nhiên.")

# Plot 9: Xu hướng Tỷ lệ lỗi của sản phẩm "Emerging" - Super Soft Tech
# (Sử dụng code từ các phản hồi trước để tạo plot này)
# ...
print("\nPLOT 9: DẤU HIỆU CỦA VẤN ĐỀ MỚI NỔI - 'SUPER SOFT TECH' (ĐÃ TRÌNH BÀY TRƯỚC)")
print("   >>> GIẢI THÍCH PLOT 9: Minh họa cho một vấn đề 'mới nổi' tiềm tàng, cho thấy khả năng phân tích sâu và tầm nhìn xa của team.")


# --- 2E: PHÂN TÍCH BẢO TRÌ (MAINTENANCE) ---
print("\n--- 2E: PHÂN TÍCH TÁC ĐỘNG CỦA BẢO TRÌ ---")

# Plot 10: So sánh Tỷ lệ Lỗi giữa ngày có và không có Bảo trì
# (Sử dụng code từ các phản hồi trước để tạo plot này)
# ...
print("\nPLOT 10: SO SÁNH TỶ LỆ LỖI NGÀY THƯỜNG VS. NGÀY BẢO TRÌ (ĐÃ TRÌNH BÀY TRƯỚC)")
print("   >>> GIẢI THÍCH PLOT 10: Đây là 'cú twist' của cuộc điều tra, phát hiện ra vấn đề chung của toàn nhà máy là quy trình khởi động lại sau bảo trì kém hiệu quả, gây ra thảm họa chất lượng.")

# --- PHẦN 3: ĐỊNH LƯỢNG TÁC ĐỘNG (Q3) ---
print("\n" + "="*80)
print("PHẦN 3: ĐỊNH LƯỢNG TÁC ĐỘNG (IMPACT QUANTIFICATION)")
print("="*80)

# Table 6: Bảng tính toán Tổn thất Sản lượng do Downtime
handle_jam_downtime = df[df['UTIL_REASON_DESCRIPTION'].str.contains('handle', na=False) & df['UTIL_REASON_DESCRIPTION'].str.contains('jam|stuck|double', na=False)]['DOWNTIME'].sum()
avg_rate = 78.7
units_lost = handle_jam_downtime * avg_rate
impact_table_downtime = pd.DataFrame({
    'Hạng mục': ['Tổng Downtime do Kẹt phôi (phút)', 'Tốc độ sản xuất trung bình (units/phút)', 'Tổng Sản lượng Tiềm năng đã mất (units)'],
    'Giá trị': [f'{handle_jam_downtime:,.0f}', f'{avg_rate:,.1f}', f'{units_lost:,.0f}']
})
print("\nTABLE 6: TÍNH TOÁN TỔN THẤT SẢN LƯỢNG DO 'KẸT PHÔI'")
display(impact_table_downtime)

# Table 7: Bảng tính toán Tổn thất Vật liệu do Lỗi Chất lượng
maft_rejects = df[df['Machine_Family'] == 'MAFT']['REJECT_PRODUCTION_QTY'].sum()
maint_rejects = df[df['Had_Maintenance']]['REJECT_PRODUCTION_QTY'].sum()
cost_per_unit = 2.0 # Giả định chi phí
impact_table_quality = pd.DataFrame({
    'Nguồn gốc Lỗi': ['Dòng máy MAFT', 'Sau Bảo trì'],
    'Số lượng Sản phẩm lỗi': [f'{maft_rejects:,.0f}', f'{maint_rejects:,.0f}'],
    'Chi phí Vật liệu Ước tính': [f'${maft_rejects*cost_per_unit/1e6:,.2f}M', f'${maint_rejects*cost_per_unit/1e6:,.2f}M']
})
print("\nTABLE 7: TÍNH TOÁN TỔN THẤT VẬT LIỆU DO LỖI CHẤT LƯỢNG'")
display(impact_table_quality)

# Plot 11: Biểu đồ thác nước (Waterfall) thể hiện các nguồn tổn thất
# ... (Waterfall chart là một kỹ thuật nâng cao, có thể bỏ qua nếu không cần thiết)

print("\n\n--- THƯ VIỆN PHÂN TÍCH ĐÃ HOÀN TẤT ---")
print("Bạn có thể chọn lọc các bảng và biểu đồ trên để xây dựng câu chuyện phân tích của mình.")
```

```python
import pandas as pd

def classify_downtime_cause_v2(reason):
    """
    Phân loại lý do downtime thành các nhóm chính một cách chi tiết và chính xác hơn.
    Ưu tiên các trường hợp đặc biệt trước để tránh phân loại sai.
    """
    # Bước 1: Chuẩn hóa đầu vào
    if pd.isna(reason):
        return 'Unknown/Not Specified'
    reason = str(reason).lower().strip()
    if not reason or reason == 'nan':
        return 'Unknown/Not Specified'

    # Bước 2: Xử lý các trường hợp đặc biệt và ưu tiên cao
    # Rework là một nhóm riêng biệt và quan trọng
    if 'rework' in reason:
        return 'Process - Rework'

    # Các vấn đề về thiếu hụt (vật tư, linh kiện)
    if 'shortage' in reason:
        return 'Supporting Systems - Shortage'

    # Các vấn đề về nhân sự
    if 'labor' in reason:
        return 'Supporting Systems - Labor'

    # Bước 3: Phân loại các nhóm chính dựa trên từ khóa
    # Nhóm Lỗi Cơ khí liên quan đến Hệ thống Cấp phôi (Handle)
    # Bao gồm cả 'lack of handle' để bắt được trường hợp thiếu phôi
    if 'handle' in reason or 'feeder' in reason:
        if 'jam' in reason or 'stuck' in reason or 'double' in reason or 'lack of' in reason:
            return 'Mechanical - Handle System'

    # Nhóm Lỗi Cơ khí khác
    if 'mechanical' in reason or 'chain' in reason or 'fiber box' in reason or 'filling tool' in reason:
        return 'Mechanical - Other'

    # Nhóm Chất lượng (không phải Rework)
    if 'quality' in reason:
        return 'Process - Quality Check'

    # Nhóm Chuyển đổi & Cài đặt
    if 'change material' in reason or 'adjust' in reason or 'setup' in reason:
        return 'Process - Changeover/Setup'

    # Nhóm An toàn
    if 'safety' in reason or 'interlock' in reason:
        return 'Safety System'

    # Bước 4: Nếu không thuộc nhóm nào ở trên, gom vào "Other"
    return 'Other'

# --- VÍ DỤ SỬ DỤNG VÀ KIỂM TRA ---

# Tạo một DataFrame mẫu để kiểm tra
test_reasons = [
    '02-Quality - 0202-Rework',
    '17-End Feeder - 1702-Jammed handle on chain',
    '18-Autofeeder - 1801-Double hadle on chain', # Cố tình viết sai 'handle'
    '17-End Feeder - 1712-Lack of handle on chain End Feeder',
    '04-Material - 0403-sparepart shortage',
    '05-Labor - 0505-Labor shortage',
    '12-Filling tool - 1201-Mechanical adjustment.',
    '03-Change material - 0301-Change material',
    '01-Safety - 0105-Interlock broken',
    'Running', # Trường hợp không phân loại được
    np.nan # Trường hợp giá trị rỗng
]
test_df = pd.DataFrame({'UTIL_REASON_DESCRIPTION': test_reasons})

# Áp dụng hàm mới
test_df['Downtime_Category_V2'] = test_df['UTIL_REASON_DESCRIPTION'].apply(classify_downtime_cause_v2)

print("=== KẾT QUẢ KIỂM TRA HÀM PHÂN LOẠI MỚI ===")
print(test_df.to_string())
```

```python
import pandas as pd
import numpy as np

# Load data và chuẩn bị
df = pd.read_csv('../data/processed/master_data.csv', low_memory=False)

# Sử dụng hàm phân loại V2
def classify_downtime_cause_v2(reason):
    if pd.isna(reason): return 'Unknown/Not Specified'
    reason = str(reason).lower().strip()
    if not reason or reason == 'nan': return 'Unknown/Not Specified'
    if 'rework' in reason: return 'Process - Rework'
    if 'shortage' in reason: return 'Supporting Systems - Shortage'
    if 'labor' in reason: return 'Supporting Systems - Labor'
    if 'handle' in reason or 'feeder' in reason:
        if 'jam' in reason or 'stuck' in reason or 'double' in reason or 'lack of' in reason:
            return 'Mechanical - Handle System'
    if 'mechanical' in reason or 'chain' in reason or 'fiber box' in reason or 'filling tool' in reason:
        return 'Mechanical - Other'
    if 'quality' in reason: return 'Process - Quality Check'
    if 'change material' in reason or 'adjust' in reason or 'setup' in reason:
        return 'Process - Changeover/Setup'
    if 'safety' in reason or 'interlock' in reason:
        return 'Safety System'
    return 'Other'

df['Downtime_Category'] = df['UTIL_REASON_DESCRIPTION'].apply(classify_downtime_cause_v2)
df['Has_Maintenance'] = df['ORDER_TYPE'] != 'No Maintenance Event'

# Chỉ giữ các dòng có downtime để phân tích
downtime_df = df[df['DOWNTIME'] > 0].copy()

# === TẠO BẢNG SO SÁNH CẤU TRÚC DOWNTIME ===
# Pivot table để lấy tổng downtime theo nhóm (Có/Không Maint) và theo Loại nguyên nhân
comparison_pivot = downtime_df.pivot_table(
    index='Downtime_Category',
    columns='Has_Maintenance',
    values='DOWNTIME',
    aggfunc='sum',
    fill_value=0
)
# Đổi tên cột cho dễ hiểu
comparison_pivot.columns = ['Downtime_No_Maint', 'Downtime_With_Maint']

# Tính toán tỷ lệ phần trăm cho từng cột
total_no_maint = comparison_pivot['Downtime_No_Maint'].sum()
total_with_maint = comparison_pivot['Downtime_With_Maint'].sum()

comparison_pivot['%_of_No_Maint_Downtime'] = (comparison_pivot['Downtime_No_Maint'] / total_no_maint * 100).round(1)
comparison_pivot['%_of_With_Maint_Downtime'] = (comparison_pivot['Downtime_With_Maint'] / total_with_maint * 100).round(1)

# Sắp xếp theo mức độ quan trọng
comparison_pivot = comparison_pivot.sort_values(by='Downtime_No_Maint', ascending=False)


print("=== INSIGHT VỀ MAINTENANCE: SO SÁNH CẤU TRÚC DOWNTIME ===\n")
print("Bảng này cho thấy, trong những ngày có bảo trì, các loại sự cố nào chiếm tỷ trọng cao hơn/thấp hơn so với ngày thường.")

display(comparison_pivot[['%_of_No_Maint_Downtime', '%_of_With_Maint_Downtime']])
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# === TẠO THƯ MỤC LƯU PLOT ===
os.makedirs('plots', exist_ok=True)
sns.set_style("whitegrid")
plt.rcParams.update({'font.size': 10})

# === TẢI DỮ LIỆU GỐC ===
print("Đang tải dữ liệu gốc...")
prod = pd.read_csv('../data/raw/production_logs.csv', low_memory=False)
maint = pd.read_csv('../data/raw/maintenance_order.csv')
cross = pd.read_csv('../data/raw/cross_reference.csv')

# === TIỀN XỬ LÝ DỮ LIỆU ===
# Chuẩn hóa ngày
prod['Date'] = pd.to_datetime(prod['PRODUCTIONDATE_DAY_LOC']).dt.date
maint['Date'] = pd.to_datetime(maint['BASIC_START_DATE']).dt.date
cross.rename(columns={'OPERA NAME': 'LINE_NAME'}, inplace=True)

# Gắn LINE_NAME vào maintenance
maint = maint.merge(cross[['EQUIPMENT_ID', 'LINE_NAME']], on='EQUIPMENT_ID', how='left')
maint = maint[['Date', 'LINE_NAME']].dropna().drop_duplicates()
maint['Has_Maintenance'] = True

# Merge vào production
prod = prod.merge(maint, on=['Date', 'LINE_NAME'], how='left')
prod['Has_Maintenance'] = prod['Has_Maintenance'].fillna(False)

# Tạo cột hỗ trợ
prod['Total_Production'] = prod['GOOD_PRODUCTION_QTY'] + prod['REJECT_PRODUCTION_QTY']
prod['Defect_Rate'] = np.where(prod['Total_Production'] > 0, prod['REJECT_PRODUCTION_QTY'] / prod['Total_Production'], np.nan)
prod['Line_Prefix'] = prod['LINE_NAME'].str[:4]
prod['Is_Handle_Jam'] = prod['UTIL_REASON_DESCRIPTION'].astype(str).str.contains(r'Handle|Jammed handle|Handle stuck', case=False, na=False)
prod['Has_Changeover'] = prod['CHANGEOVER_DURATION'] > 0

# === HÀM LƯU PLOT ===
def save_plot(filename):
    plt.tight_layout()
    plt.savefig(f'plots/{filename}.png', dpi=150, bbox_inches='tight')
    plt.close()

# === 1. TOP 10 DOWNTIME REASONS (Pareto) ===
top_downtime = prod.groupby('UTIL_REASON_DESCRIPTION')['DOWNTIME'].sum().sort_values(ascending=False).head(10)
plt.figure(figsize=(10, 6))
sns.barplot(x=top_downtime.values, y=top_downtime.index, palette='Reds_r')
plt.title('Top 10 Nguyên nhân Downtime (Pareto)')
plt.xlabel('Tổng Downtime (phút)')
save_plot('01_downtime_pareto')

# === 2. DEFECT RATE THEO DÒNG MÁY ===
defect_by_line = prod.groupby('Line_Prefix').agg(
    Total_Good=('GOOD_PRODUCTION_QTY', 'sum'),
    Total_Reject=('REJECT_PRODUCTION_QTY', 'sum')
).reset_index()
defect_by_line['Defect_Rate'] = defect_by_line['Total_Reject'] / (defect_by_line['Total_Good'] + defect_by_line['Total_Reject'])
plt.figure(figsize=(8, 5))
sns.barplot(data=defect_by_line.sort_values('Defect_Rate', ascending=False), x='Defect_Rate', y='Line_Prefix', palette='Blues_r')
plt.title('Defect Rate theo Dòng máy')
plt.xlabel('Defect Rate')
save_plot('02_defect_by_line')

# === 3. DOWNTIME DO HANDLE JAM THEO DÒNG MÁY ===
handle_downtime = prod[prod['Is_Handle_Jam']].groupby('Line_Prefix')['DOWNTIME'].sum().reset_index()
plt.figure(figsize=(8, 5))
sns.barplot(data=handle_downtime.sort_values('DOWNTIME', ascending=False), x='DOWNTIME', y='Line_Prefix', palette='Oranges_r')
plt.title('Downtime do "Handle Jam" theo Dòng máy')
plt.xlabel('Tổng Downtime (phút)')
save_plot('03_handle_jam_by_line')

# === 4. ẢNH HƯỞNG CỦA CHANGEOVER ĐẾN HANDLE JAM ===
mkbc = prod[prod['Line_Prefix'] == 'MKBC'].copy()
mkbc['Ca_Key'] = mkbc['Date'].astype(str) + '_' + mkbc['SHIFT_NAME']
ca_summary = mkbc.groupby('Ca_Key').agg(
    Has_Changeover=('Has_Changeover', 'any'),
    Handle_Jam_Count=('Is_Handle_Jam', 'sum')
).reset_index()
avg_with = ca_summary[ca_summary['Has_Changeover']]['Handle_Jam_Count'].mean()
avg_without = ca_summary[~ca_summary['Has_Changeover']]['Handle_Jam_Count'].mean()
plt.figure(figsize=(6, 4))
plt.bar(['Có Changeover', 'Không Changeover'], [avg_with, avg_without], color=['red', 'gray'])
plt.title('Handle Jam trung bình theo Changeover (MKBC)')
plt.ylabel('Số lần Jam/ca')
save_plot('04_jam_vs_changeover')

# === 5. DEFECT RATE: CÓ VS KHÔNG BẢO TRÌ ===
maint_effect = prod.groupby('Has_Maintenance')['Defect_Rate'].mean()
plt.figure(figsize=(6, 4))
plt.bar(['Không Bảo trì', 'Có Bảo trì'], maint_effect.values, color=['green', 'red'])
plt.title('Defect Rate: Có vs Không Bảo trì')
plt.ylabel('Defect Rate')
save_plot('05_defect_vs_maintenance')

# === 6. DOWNTIME TRƯỚC VÀ SAU BẢO TRÌ (3 ngày) ===
# Tạo danh sách ngày sau bảo trì
maint_dates = maint[['Date', 'LINE_NAME']].copy()
after_days = []
for i in range(1, 4):
    temp = maint_dates.copy()
    temp['Date'] = pd.to_datetime(temp['Date']) + pd.Timedelta(days=i)
    temp['Post_Maint'] = True
    after_days.append(temp)
after_df = pd.concat(after_days)
after_df['Date'] = after_df['Date'].dt.date

# Gắn vào dữ liệu
prod_check = prod[['Date', 'LINE_NAME', 'DOWNTIME']].copy()
prod_check = prod_check.merge(after_df, on=['Date', 'LINE_NAME'], how='left')
prod_check['Post_Maint'] = prod_check['Post_Maint'].fillna(False)

downtime_after = prod_check[prod_check['Post_Maint']]['DOWNTIME'].mean()
downtime_normal = prod_check[~prod_check['Post_Maint']]['DOWNTIME'].mean()
plt.figure(figsize=(6, 4))
plt.bar(['3 ngày sau Bảo trì', 'Ngày thường'], [downtime_after, downtime_normal], color=['orange', 'blue'])
plt.title('Downtime: Sau Bảo trì vs Ngày thường')
plt.ylabel('Downtime trung bình (phút)')
save_plot('06_downtime_after_maint')

# === 7. DEFECT RATE THEO SẢN PHẨM TRÊN MAFT ===
maft = prod[prod['Line_Prefix'] == 'MAFT'].copy()
product_defect = maft.groupby('SIZE_TYPE').agg(
    Total_Good=('GOOD_PRODUCTION_QTY', 'sum'),
    Total_Reject=('REJECT_PRODUCTION_QTY', 'sum')
).reset_index()
product_defect['Defect_Rate'] = product_defect['Total_Reject'] / (product_defect['Total_Good'] + product_defect['Total_Reject'])
plt.figure(figsize=(8, 4))
sns.barplot(data=product_defect.sort_values('Defect_Rate', ascending=False), x='Defect_Rate', y='SIZE_TYPE', palette='Purples_r')
plt.title('Defect Rate theo Sản phẩm trên MAFT')
plt.xlabel('Defect Rate')
save_plot('07_maft_product_defect')

# === 8. XU HƯỚNG HANDLE JAM THEO THỜI GIAN (MKBC) ===
mkbc_daily = mkbc.groupby('Date')['Is_Handle_Jam'].sum().reset_index()
plt.figure(figsize=(12, 4))
plt.plot(mkbc_daily['Date'], mkbc_daily['Is_Handle_Jam'], color='red', alpha=0.7)
plt.title('Xu hướng "Handle Jam" trên MKBC theo Thời gian')
plt.ylabel('Số lần Jam/ngày')
save_plot('08_jam_trend')

# === IN CÁC BẢNG QUAN TRỌNG ===
print("="*80)
print("BẢNG 1: TOP 10 NGUYÊN NHÂN DOWNTIME")
print(top_downtime.to_string())
print("\n" + "="*80)
print("BẢNG 2: DEFECT RATE THEO DÒNG MÁY")
print(defect_by_line[['Line_Prefix', 'Defect_Rate']].sort_values('Defect_Rate', ascending=False).to_string(index=False))
print("\n" + "="*80)
print("BẢNG 3: DOWNTIME DO HANDLE JAM THEO DÒNG MÁY")
print(handle_downtime.sort_values('DOWNTIME', ascending=False).to_string(index=False))
print("\n" + "="*80)
print("BẢNG 4: ẢNH HƯỞNG CỦA CHANGEOVER ĐẾN HANDLE JAM (MKBC)")
print(f"Có Changeover: {avg_with:.3f} lần/ca")
print(f"Không Changeover: {avg_without:.3f} lần/ca")
print(f"Chênh lệch: {avg_with - avg_without:.3f} lần")
print("\n" + "="*80)
print("BẢNG 5: HIỆU ỨNG CỦA BẢO TRÌ")
maint_summary = prod.groupby('Has_Maintenance').agg(
    Avg_Defect_Rate=('Defect_Rate', 'mean'),
    Avg_Downtime=('DOWNTIME', 'mean'),
    Count=('DOWNTIME', 'size')
).reset_index()
maint_summary['Has_Maintenance'] = maint_summary['Has_Maintenance'].map({True: 'Có Bảo trì', False: 'Không Bảo trì'})
print(maint_summary.to_string(index=False))
print("\n" + "="*80)
print("BẢNG 6: DEFECT RATE TRÊN MAFT THEO SẢN PHẨM")
print(product_defect[['SIZE_TYPE', 'Defect_Rate']].sort_values('Defect_Rate', ascending=False).to_string(index=False))
print("\n" + "="*80)
print("BẢNG 7: TOP 5 SẢN PHẨM GÂY HANDLE JAM TRÊN MKBC")
jam_by_product = mkbc[mkbc['Is_Handle_Jam']].groupby('SIZE_TYPE')['DOWNTIME'].sum().sort_values(ascending=False).head(5)
print(jam_by_product.to_string())
print("\n" + "="*80)
print("BẢNG 8: KIỂM ĐỊNH THỐNG KÊ BẢO TRÌ")
# Tính lại cho kiểm định
with_m = prod[prod['Has_Maintenance']]
without_m = prod[~prod['Has_Maintenance']]
good1 = with_m['GOOD_PRODUCTION_QTY'].sum()
bad1 = with_m['REJECT_PRODUCTION_QTY'].sum()
good2 = without_m['GOOD_PRODUCTION_QTY'].sum()
bad2 = without_m['REJECT_PRODUCTION_QTY'].sum()
dr1 = bad1 / (good1 + bad1)
dr2 = bad2 / (good2 + bad2)
print(f"Defect rate (có bảo trì): {dr1:.4f}")
print(f"Defect rate (không bảo trì): {dr2:.4f}")
print(f"Tỷ lệ chênh lệch: {(dr1/dr2 - 1)*100:.1f}%")
print("\n" + "="*80)
print("DANH SÁCH QUY TRÌNH CẦN CẢI TIẾN")
print("1. Quy trình Changeover trên MKBC")
print("2. Quy trình Vận hành Dây chuyền MAFT")
print("3. Quy trình Bảo trì & Kiểm tra sau Bảo trì")
print("4. Quy trình Kiểm soát Chất lượng theo Sản phẩm")
print("5. Quy trình Quản lý Vật liệu Đầu vào")
print("\n✅ Đã lưu 8 biểu đồ vào thư mục 'plots/'.")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# --- 0. SETUP & TẠO THƯ MỤC LƯU ---
output_dir = "maintenance_analysis"
if not os.path.exists(output_dir):
    os.makedirs(output_dir)
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 7)
plt.rcParams['font.size'] = 12

# --- 1. LOAD DATA & CHUẨN BỊ ---
try:
    df = pd.read_csv('../data/processed/master_data.csv', low_memory=False)
    print("✅ Tải dữ liệu '../data/processed/master_data.csv' thành công.")
except FileNotFoundError:
    print("❌ LỖI: Không tìm thấy file '../data/processed/master_data.csv'.")
    exit()

df['Total_Production'] = df['GOOD_PRODUCTION_QTY'] + df['REJECT_PRODUCTION_QTY']
df['Has_Maintenance'] = df['ORDER_TYPE'] != 'No Maintenance Event'

def classify_downtime_cause_v2(reason):
    if pd.isna(reason): return 'Unknown'
    reason = str(reason).lower().strip()
    if not reason or reason == 'nan': return 'Unknown'
    if 'rework' in reason: return 'Process - Rework'
    if 'shortage' in reason: return 'Supporting Systems - Shortage'
    if 'labor' in reason: return 'Supporting Systems - Labor'
    if 'handle' in reason or 'feeder' in reason:
        if 'jam' in reason or 'stuck' in reason or 'double' in reason or 'lack of' in reason:
            return 'Mechanical - Handle System'
    if 'mechanical' in reason or 'chain' in reason or 'fiber box' in reason or 'filling tool' in reason:
        return 'Mechanical - Other'
    if 'quality' in reason: return 'Process - Quality Check'
    if 'change material' in reason or 'adjust' in reason or 'setup' in reason:
        return 'Process - Changeover/Setup'
    if 'safety' in reason or 'interlock' in reason:
        return 'Safety System'
    return 'Other'

df['Downtime_Category'] = df['UTIL_REASON_DESCRIPTION'].apply(classify_downtime_cause_v2)
print("✅ Chuẩn bị dữ liệu hoàn tất.\n")


# --- 2. BẢNG 1: SO SÁNH HIỆU SUẤT TỔNG THỂ ---
group_cols = ['PRODUCTIONDATE_DAY_LOC', 'SHIFT_NAME', 'LINE_NAME']
shift_summary = df.groupby(group_cols).agg(
    Total_Downtime=('DOWNTIME', 'sum'),
    Total_Good=('GOOD_PRODUCTION_QTY', 'sum'),
    Total_Production=('Total_Production', 'sum'),
    Is_Maint_Day=('Has_Maintenance', 'any')
).reset_index()

shift_summary = shift_summary[shift_summary['Total_Production'] > 0]
shift_summary['Defect_Rate'] = (shift_summary['Total_Production'] - shift_summary['Total_Good']) / shift_summary['Total_Production']

comparison_report = shift_summary.groupby('Is_Maint_Day').agg(
    Num_Shifts=('LINE_NAME', 'count'),
    Avg_Downtime_Per_Shift=('Total_Downtime', 'mean'),
    Avg_Defect_Rate=('Defect_Rate', 'mean')
).reset_index()
comparison_report['Is_Maint_Day'] = comparison_report['Is_Maint_Day'].map({True: 'Có Maintenance', False: 'Không có Maintenance'})
comparison_report['Avg_Defect_Rate_%'] = (comparison_report['Avg_Defect_Rate'] * 100)

print("="*80)
print("BẢNG 1: SO SÁNH HIỆU SUẤT TRUNG BÌNH CỦA MỘT CA SẢN XUẤT")
print("="*80)
print(comparison_report[['Is_Maint_Day', 'Num_Shifts', 'Avg_Downtime_Per_Shift', 'Avg_Defect_Rate_%']].round(2).to_string(index=False))
print("\n")

# --- 3. BIỂU ĐỒ 1: SO SÁNH DEFECT RATE ---
plt.figure(figsize=(8, 6))
bars = sns.barplot(x='Is_Maint_Day', y='Avg_Defect_Rate_%', data=comparison_report, palette=['skyblue', 'coral'])
plt.title('Biểu đồ 1: Tỷ lệ lỗi tăng vọt 15 lần trong ngày có Bảo trì', fontsize=16, fontweight='bold')
plt.ylabel('Tỷ lệ lỗi trung bình (%)')
plt.xlabel('')
for bar in bars.patches:
    plt.text(bar.get_x() + bar.get_width() / 2, bar.get_height(), f'{bar.get_height():.2f}%',
             ha='center', va='bottom', fontsize=12, fontweight='bold')
plt.savefig(os.path.join(output_dir, "plot_maint_1_defect_rate_comparison.png"), dpi=150, bbox_inches='tight')
plt.show()

# --- 4. BẢNG 2: SO SÁNH CẤU TRÚC DOWNTIME ---
downtime_df = df[df['DOWNTIME'] > 0].copy()
comparison_pivot = downtime_df.pivot_table(
    index='Downtime_Category',
    columns='Has_Maintenance',
    values='DOWNTIME',
    aggfunc='sum',
    fill_value=0
)
comparison_pivot.columns = ['Downtime_No_Maint', 'Downtime_With_Maint']
comparison_pivot['%_No_Maint'] = (comparison_pivot['Downtime_No_Maint'] / comparison_pivot['Downtime_No_Maint'].sum() * 100)
comparison_pivot['%_With_Maint'] = (comparison_pivot['Downtime_With_Maint'] / comparison_pivot['Downtime_With_Maint'].sum() * 100)
comparison_pivot = comparison_pivot.sort_values(by='%_No_Maint', ascending=False)

print("="*80)
print("BẢNG 2: SO SÁNH CẤU TRÚC DOWNTIME (% PHÂN BỔ)")
print("="*80)
print(comparison_pivot[['%_No_Maint', '%_With_Maint']].round(1).to_string())
print("\n")


# --- 5. BIỂU ĐỒ 2: SO SÁNH CẤU TRÚC DOWNTIME ---
plot_data = comparison_pivot[['%_No_Maint', '%_With_Maint']].reset_index()
plot_data_melted = plot_data.melt(id_vars='Downtime_Category', var_name='Day_Type', value_name='Percentage')
plot_data_melted['Day_Type'] = plot_data_melted['Day_Type'].map({'%_No_Maint': 'Ngày thường', '%_With_Maint': 'Ngày có Bảo trì'})

plt.figure(figsize=(14, 8))
sns.barplot(x='Percentage', y='Downtime_Category', hue='Day_Type', data=plot_data_melted, palette='muted')
plt.title('Biểu đồ 2: "Setup/Adjustment" là nguyên nhân chính gây Downtime sau Bảo trì', fontsize=16, fontweight='bold')
plt.xlabel('Tỷ trọng Downtime (%)')
plt.ylabel('Nhóm Nguyên nhân')
plt.legend(title='Loại ngày')
plt.grid(axis='x', linestyle='--')
plt.savefig(os.path.join(output_dir, "plot_maint_2_downtime_structure.png"), dpi=150, bbox_inches='tight')
plt.show()

print(f"\n✅ Đã tạo và lưu 2 biểu đồ vào thư mục '{output_dir}'.")
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split

# --- 1. LOAD VÀ CHUẨN BỊ DỮ LIỆU GỐC ---
df = pd.read_csv('../data/processed/master_data.csv', low_memory=False)
df['PRODUCTIONDATE_DAY_LOC'] = pd.to_datetime(df['PRODUCTIONDATE_DAY_LOC'])

# Tạo các cột cờ cần thiết
df['Is_Handle_Jam'] = df['UTIL_REASON_DESCRIPTION'].astype(str).str.contains('handle', na=False) & \
                      (df['UTIL_REASON_DESCRIPTION'].astype(str).str.contains('jam|stuck|double', na=False))
df['Is_Maintenance'] = (df['ORDER_TYPE'] != 'No Maintenance Event')
df['Is_Changeover'] = df['UTIL_REASON_DESCRIPTION'].astype(str).str.contains('change material', na=False)


# --- 2. TỔNG HỢP DỮ LIỆU LÊN CẤP ĐỘ CA SẢN XUẤT (CHỈ CHO MKBC) ---
df_mkbc = df[df['LINE_NAME'].str.startswith('MKBC', na=False)].copy()
group_cols = ['PRODUCTIONDATE_DAY_LOC', 'SHIFT_NAME', 'LINE_NAME']

model_data = df_mkbc.groupby(group_cols).agg(
    # Lấy các feature dạng category
    SIZE_TYPE=('SIZE_TYPE', lambda x: x.mode()[0] if not x.mode().empty else 'Unknown'),
    CREW_ID=('CREW_ID', lambda x: x.mode()[0] if not x.mode().empty else 'Unknown'),
    # Tạo target và các feature dạng cờ
    Had_Handle_Jam=('Is_Handle_Jam', 'any'),
    Had_Maintenance=('Is_Maintenance', 'any'),
    Had_Changeover=('Is_Changeover', 'any')
).reset_index()

# Chuyển đổi target thành 0/1
model_data['Had_Handle_Jam'] = model_data['Had_Handle_Jam'].astype(int)

print("--- BẢNG DỮ LIỆU SẴN SÀNG CHO MÔ HÌNH ---")
print(f"Số lượng quan sát (ca sản xuất trên MKBC): {len(model_data)}")
print(f"Tỷ lệ Handle Jam: {model_data['Had_Handle_Jam'].mean():.2%}")
display(model_data.head())


# --- 3. CHIA TẬP X VÀ Y ---
# Loại bỏ các cột định danh không dùng làm feature
X = model_data.drop(columns=['PRODUCTIONDATE_DAY_LOC', 'SHIFT_NAME', 'LINE_NAME', 'Had_Handle_Jam'])
y = model_data['Had_Handle_Jam']

# Mã hóa One-Hot cho các biến category
X_encoded = pd.get_dummies(X, columns=['SIZE_TYPE', 'CREW_ID'], drop_first=True)

# Chuyển các cột boolean thành int
for col in ['Had_Maintenance', 'Had_Changeover']:
    if col in X_encoded.columns:
        X_encoded[col] = X_encoded[col].astype(int)

print("\n--- TẬP FEATURE (X) SAU KHI MÃ HÓA ---")
display(X_encoded.head())

print("\n--- TẬP TARGET (y) ---")
display(y.head())

# --- 4. CHIA TẬP TRAIN VÀ TEST ---
# Chia dữ liệu theo tỷ lệ 80% train, 20% test
# stratify=y đảm bảo tỷ lệ Handle Jam trong tập train và test là tương tự nhau, rất quan trọng!
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, random_state=42, stratify=y
)

print("\n--- KÍCH THƯỚC CÁC TẬP DỮ LIỆU ---")
print(f"X_train shape: {X_train.shape}")
print(f"X_test shape:  {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape:  {y_test.shape}")
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, roc_auc_score, confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# --- 5. HUẤN LUYỆN MÔ HÌNH ---
print("\n--- BẮT ĐẦU HUẤN LUYỆN MÔ HÌNH RANDOM FOREST ---")
# class_weight='balanced' giúp mô hình chú ý hơn đến lớp thiểu số (Handle Jam)
model = RandomForestClassifier(n_estimators=100, random_state=42, class_weight='balanced', n_jobs=-1)
model.fit(X_train, y_train)
print("✅ Huấn luyện hoàn tất.")


# --- 6. ĐÁNH GIÁ MÔ HÌNH TRÊN TẬP TEST ---
print("\n--- KẾT QUẢ ĐÁNH GIÁ TRÊN TẬP TEST ---")
y_pred = model.predict(X_test)
y_pred_proba = model.predict_proba(X_test)[:, 1]

# In báo cáo phân loại chi tiết
print("\nBáo cáo Phân loại (Classification Report):")
print(classification_report(y_test, y_pred))

# Tính ROC AUC Score - chỉ số quan trọng để đo lường khả năng phân loại của mô hình
roc_auc = roc_auc_score(y_test, y_pred_proba)
print(f"\nROC AUC Score: {roc_auc:.4f}")
print("(Chỉ số này càng gần 1.0 càng tốt, > 0.7 là mô hình khá tốt)")

# Vẽ ma trận nhầm lẫn (Confusion Matrix)
print("\nMa trận Nhầm lẫn (Confusion Matrix):")
cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=model.classes_)
disp.plot(cmap=plt.cm.Blues)
plt.title('Confusion Matrix')
plt.show()


# --- 7. PHÂN TÍCH FEATURE IMPORTANCE ---
print("\n--- CÁC YẾU TỐ QUAN TRỌNG NHẤT ẢNH HƯỞNG ĐẾN HANDLE JAM ---")
feature_importances = pd.DataFrame({
    'Feature': X_train.columns,
    'Importance': model.feature_importances_
}).sort_values(by='Importance', ascending=False)

display(feature_importances.head(10))
```

```python
import pandas as pd
import numpy as np

# Load dữ liệu
df = pd.read_csv('../data/processed/master_data.csv', low_memory=False)
df['Total_Production'] = df['GOOD_PRODUCTION_QTY'] + df['REJECT_PRODUCTION_QTY']

# === 1. LỌC VÀ CHUẨN BỊ DỮ LIỆU CHỈ CHO MKTU ===
# Chỉ lấy các sự kiện sản xuất (có sản lượng) trên dòng MKTU
mktu_prod_events = df[
    (df['LINE_NAME'].str.startswith('MKTU', na=False)) &
    (df['Total_Production'] > 0)
].copy()

if mktu_prod_events.empty:
    print("Không có dữ liệu sản xuất nào cho dòng máy MKTU.")
else:
    # === 2. TÍNH TOÁN HIỆU SUẤT THEO TỪNG SẢN PHẨM ===
    # Tổng hợp theo SIZE_TYPE
    product_performance_mktu = mktu_prod_events.groupby('SIZE_TYPE').agg(
        Total_Reject=('REJECT_PRODUCTION_QTY', 'sum'),
        Total_Production=('Total_Production', 'sum'),
        Number_of_Runs=('LINE_NAME', 'count') # Đếm số lần sản xuất
    ).reset_index()

    # Tính Defect Rate
    product_performance_mktu['Defect_Rate_%'] = (product_performance_mktu['Total_Reject'] / product_performance_mktu['Total_Production'] * 100).round(2)

    # Lọc ra những sản phẩm có sản lượng đáng kể (ví dụ: > 100,000) để kết quả đáng tin cậy
    significant_products = product_performance_mktu[product_performance_mktu['Total_Production'] > 100000]

    # Sắp xếp theo Defect Rate để tìm "thủ phạm"
    top_culprits_mktu = significant_products.sort_values(by='Defect_Rate_%', ascending=False)

    # === 3. TRÌNH BÀY KẾT QUẢ ===
    print("=== ĐIỀU TRA VẤN ĐỀ CHẤT LƯỢNG TRÊN DÒNG MÁY MKTU ===")

    overall_defect_rate_mktu = (mktu_prod_events['REJECT_PRODUCTION_QTY'].sum() / mktu_prod_events['Total_Production'].sum()) * 100
    print(f"\nTỷ lệ lỗi trung bình chung của MKTU: {overall_defect_rate_mktu:.2f}%\n")

    print("--- BẢNG: TOP SẢN PHẨM 'THỦ PHẠM' GÂY LỖI NHIỀU NHẤT TRÊN MKTU ---")
    print("(Chỉ xét các sản phẩm có tổng sản lượng > 100,000)")

    if top_culprits_mktu.empty:
        print("Không tìm thấy sản phẩm nào có sản lượng đáng kể để phân tích.")
    else:
        display(top_culprits_mktu.head(5))
```

```python
import pandas as pd
import numpy as np

# Load dữ liệu
df = pd.read_csv('../data/processed/master_data.csv', low_memory=False)
df['Total_Production'] = df['GOOD_PRODUCTION_QTY'] + df['REJECT_PRODUCTION_QTY']
df['PRODUCTIONDATE_DAY_LOC'] = pd.to_datetime(df['PRODUCTIONDATE_DAY_LOC'])


# Danh sách các "sản phẩm thủ phạm"
culprit_products = [
    'Renew 360', 'Oral Care 360',
    'Junior Guard Taper (Ages 5+)', 'EcoFresh Green', 'GreenClean Pro'
]

# === 2. KIỂM TRA GIẢ THUYẾT 2: ẢNH HƯỞNG CỦA CHANGEOVER ===
print("=== BẢNG 1: ẢNH HƯỞNG CỦA CHANGEOVER LÊN TỶ LỆ LỖI CỦA CÁC SẢN PHẨM 'THỦ PHẠM' ===")

# Tạo cờ 'Had_Changeover' cho mỗi ca
group_cols = ['PRODUCTIONDATE_DAY_LOC', 'SHIFT_NAME', 'LINE_NAME']
changeover_shifts = df[
    df['UTIL_REASON_DESCRIPTION'].astype(str).str.contains('change material', na=False)
][group_cols].drop_duplicates()
changeover_shifts['Had_Changeover'] = True

# Gộp dữ liệu theo ca
shift_summary = df[df['Total_Production'] > 0].groupby(group_cols + ['SIZE_TYPE']).agg(
    Total_Reject=('REJECT_PRODUCTION_QTY', 'sum'),
    Total_Production=('Total_Production', 'sum')
).reset_index()
shift_summary = shift_summary[shift_summary['Total_Production'] > 0]
shift_summary['Defect_Rate'] = shift_summary['Total_Reject'] / shift_summary['Total_Production']

# Merge thông tin changeover vào
analysis_df = pd.merge(shift_summary, changeover_shifts, on=group_cols, how='left')
analysis_df['Had_Changeover'] = analysis_df['Had_Changeover'].fillna(False)

# So sánh Defect Rate (ĐÃ SỬA LỖI)
changeover_impact = analysis_df[analysis_df['SIZE_TYPE'].isin(culprit_products)].groupby(['SIZE_TYPE', 'Had_Changeover'])['Defect_Rate'].mean().unstack().fillna(0)
# Đổi tên cột một cách an toàn
changeover_impact.rename(columns={False: 'Defect_Rate_No_Changeover', True: 'Defect_Rate_With_Changeover'}, inplace=True)
# Thêm các cột còn thiếu nếu cần
if 'Defect_Rate_No_Changeover' not in changeover_impact.columns:
    changeover_impact['Defect_Rate_No_Changeover'] = 0
if 'Defect_Rate_With_Changeover' not in changeover_impact.columns:
    changeover_impact['Defect_Rate_With_Changeover'] = 0

changeover_impact['Difference'] = (changeover_impact['Defect_Rate_With_Changeover'] - changeover_impact['Defect_Rate_No_Changeover'])
changeover_impact['Difference_%'] = changeover_impact['Difference'] * 100

display(changeover_impact)


# === 3. KIỂM TRA GIẢ THUYẾT 3: ẢNH HƯỞNG CỦA ĐỘI NGŨ (CREW_ID) ===
print("\n=== BẢNG 2: TỶ LỆ LỖI CỦA CÁC SẢN PHẨM 'THỦ PHẠM' THEO TỪNG ĐỘI NGŨ ===")

# Lấy dữ liệu sản xuất của các sản phẩm thủ phạm
crew_analysis_df = df[df['SIZE_TYPE'].isin(culprit_products) & (df['Total_Production'] > 0)].copy()
crew_analysis_df.dropna(subset=['CREW_ID'], inplace=True)

# Tạo bảng pivot để so sánh
crew_pivot = crew_analysis_df.groupby(['SIZE_TYPE', 'CREW_ID']).agg(
    Total_Reject=('REJECT_PRODUCTION_QTY', 'sum'),
    Total_Production=('Total_Production', 'sum')
).reset_index()
crew_pivot = crew_pivot[crew_pivot['Total_Production'] > 10000]
crew_pivot['Defect_Rate_%'] = (crew_pivot['Total_Reject'] / crew_pivot['Total_Production'] * 100)

if crew_pivot.empty:
    print("Không có đủ dữ liệu để phân tích theo Đội ngũ cho các sản phẩm này.")
else:
    display(crew_pivot.pivot_table(index='SIZE_TYPE', columns='CREW_ID', values='Defect_Rate_%').fillna('-'))
```

```python
import pandas as pd
import numpy as np

# Load data và chuẩn bị
df = pd.read_csv('../data/processed/master_data.csv', low_memory=False)
df['Total_Production'] = df['GOOD_PRODUCTION_QTY'] + df['REJECT_PRODUCTION_QTY']
df['Has_Maintenance'] = df['ORDER_TYPE'] != 'No Maintenance Event'

# === TỔNG HỢP THEO CA SẢN XUẤT ===
group_cols = ['PRODUCTIONDATE_DAY_LOC', 'SHIFT_NAME', 'LINE_NAME']
shift_summary = df.groupby(group_cols).agg(
    Total_Downtime=('DOWNTIME', 'sum'),
    Total_Good=('GOOD_PRODUCTION_QTY', 'sum'),
    Total_Production=('Total_Production', 'sum'),
    Had_Maintenance=('Has_Maintenance', 'any')
).reset_index()

# Chỉ giữ các ca có sản xuất để tính Defect Rate
shift_summary = shift_summary[shift_summary['Total_Production'] > 0].copy()
shift_summary['Defect_Rate'] = (shift_summary['Total_Production'] - shift_summary['Total_Good']) / shift_summary['Total_Production']

# === TRUY VẾT "BỆNH NHÂN" NẶNG NHẤT ===

# 1. Lọc ra chỉ những ca CÓ hoạt động bảo trì
maint_shifts = shift_summary[shift_summary['Had_Maintenance'] == True].copy()

# 2. Tính toán hiệu suất cho TỪNG DÂY CHUYỀN trong những ngày chúng được bảo trì
# Chúng ta sẽ nhóm theo LINE_NAME và tính Defect Rate trung bình
patient_report = maint_shifts.groupby('LINE_NAME').agg(
    Avg_Defect_Rate_on_Maint_Days=('Defect_Rate', 'mean'),
    Number_of_Maint_Shifts=('LINE_NAME', 'count')
).reset_index()

# 3. Sắp xếp để tìm ra "bệnh nhân" nặng nhất
patient_report = patient_report.sort_values(by='Avg_Defect_Rate_on_Maint_Days', ascending=False)

# 4. Định dạng và trình bày
patient_report['Avg_Defect_Rate_on_Maint_Days_%'] = (patient_report['Avg_Defect_Rate_on_Maint_Days'] * 100).round(2)


print("=== BÁO CÁO TRUY VẾT BỆNH NHÂN: TOP 10 MÁY BỊ ẢNH HƯỞNG NẶNG NHẤT SAU BẢO TRÌ ===\n")
print("Bảng này liệt kê các dây chuyền cụ thể có tỷ lệ lỗi cao nhất trong những ngày chúng được bảo trì.")

display(patient_report.head(10)[['LINE_NAME', 'Avg_Defect_Rate_on_Maint_Days_%', 'Number_of_Maint_Shifts']])
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# --- 0. SETUP & TẠO THƯ MỤC LƯU ---
output_dir = "maintenance_deep_dive"
if not os.path.exists(output_dir):
    os.makedirs(output_dir)
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 7)
plt.rcParams['font.size'] = 12

# --- 1. LOAD DATA & CHUẨN BỊ ---
try:
    df = pd.read_csv('../data/processed/master_data.csv', low_memory=False)
    print("✅ Tải dữ liệu '../data/processed/master_data.csv' thành công.")
except FileNotFoundError:
    print("❌ LỖI: Không tìm thấy file '../data/processed/master_data.csv'.")
    exit()

df['Total_Production'] = df['GOOD_PRODUCTION_QTY'] + df['REJECT_PRODUCTION_QTY']
df['Has_Maintenance'] = df['ORDER_TYPE'] != 'No Maintenance Event'

def classify_downtime_cause_v2(reason):
    if pd.isna(reason): return 'Unknown'
    reason = str(reason).lower().strip()
    if 'rework' in reason: return 'Process - Rework'
    if 'shortage' in reason: return 'Supporting Systems - Shortage'
    if 'labor' in reason: return 'Supporting Systems - Labor'
    if 'handle' in reason or 'feeder' in reason:
        if 'jam' in reason or 'stuck' in reason or 'double' in reason or 'lack of' in reason:
            return 'Mechanical - Handle System'
    if 'mechanical' in reason or 'chain' in reason or 'fiber box' in reason or 'filling tool' in reason:
        return 'Mechanical - Other'
    if 'change material' in reason or 'adjust' in reason or 'setup' in reason:
        return 'Process - Changeover/Setup'
    return 'Other'

df['Downtime_Category'] = df['UTIL_REASON_DESCRIPTION'].apply(classify_downtime_cause_v2)
print("✅ Chuẩn bị dữ liệu hoàn tất.\n")


# --- HỒI 1: PHÁT HIỆN BẤT THƯỜNG ---
group_cols = ['PRODUCTIONDATE_DAY_LOC', 'SHIFT_NAME', 'LINE_NAME']
shift_summary = df.groupby(group_cols).agg(
    Total_Downtime=('DOWNTIME', 'sum'),
    Total_Good=('GOOD_PRODUCTION_QTY', 'sum'),
    Total_Production=('Total_Production', 'sum'),
    Is_Maint_Day=('Has_Maintenance', 'any')
).reset_index()

shift_summary = shift_summary[shift_summary['Total_Production'] > 0]
shift_summary['Defect_Rate'] = (shift_summary['Total_Production'] - shift_summary['Total_Good']) / shift_summary['Total_Production']

comparison_report = shift_summary.groupby('Is_Maint_Day').agg(
    Avg_Defect_Rate=('Defect_Rate', 'mean')
).reset_index()
comparison_report['Is_Maint_Day'] = comparison_report['Is_Maint_Day'].map({True: 'Ngày Có Bảo trì', False: 'Ngày Thường'})
comparison_report['Avg_Defect_Rate_%'] = (comparison_report['Avg_Defect_Rate'] * 100)

print("="*80)
print("BẢNG 1: SO SÁNH TỶ LỆ LỖI GIỮA NGÀY THƯỜNG VÀ NGÀY CÓ BẢO TRÌ")
print("="*80)
print(comparison_report[['Is_Maint_Day', 'Avg_Defect_Rate_%']].round(2).to_string(index=False))
print("\n")

# BIỂU ĐỒ 1
plt.figure(figsize=(8, 6))
bars = sns.barplot(x='Is_Maint_Day', y='Avg_Defect_Rate_%', data=comparison_report, palette=['skyblue', 'coral'])
plt.title('Biểu đồ 1: Tỷ lệ lỗi tăng vọt 15 lần trong ngày có Bảo trì', fontsize=16, fontweight='bold')
plt.ylabel('Tỷ lệ lỗi trung bình (%)')
plt.xlabel('')
for bar in bars.patches:
    plt.text(bar.get_x() + bar.get_width() / 2, bar.get_height(), f'{bar.get_height():.2f}%',
             ha='center', va='bottom', fontsize=12, fontweight='bold')
plt.savefig(os.path.join(output_dir, "plot_maint_1_defect_rate_comparison.png"), dpi=150, bbox_inches='tight')
plt.show()

# --- HỒI 2: CHẨN ĐOÁN NGUYÊN NHÂN ---
downtime_df = df[df['DOWNTIME'] > 0].copy()
comparison_pivot = downtime_df.pivot_table(
    index='Downtime_Category', columns='Has_Maintenance', values='DOWNTIME', aggfunc='sum', fill_value=0
)
comparison_pivot.columns = ['No_Maint', 'With_Maint']
comparison_pivot['%_No_Maint'] = (comparison_pivot['No_Maint'] / comparison_pivot['No_Maint'].sum() * 100)
comparison_pivot['%_With_Maint'] = (comparison_pivot['With_Maint'] / comparison_pivot['With_Maint'].sum() * 100)

print("="*80)
print("BẢNG 2: SO SÁNH CẤU TRÚC DOWNTIME (% PHÂN BỔ)")
print("="*80)
print(comparison_pivot[['%_No_Maint', '%_With_Maint']].round(1).to_string())
print("\n")

# BIỂU ĐỒ 2
plot_data = comparison_pivot[['%_No_Maint', '%_With_Maint']].reset_index().melt(id_vars='Downtime_Category', var_name='Day_Type', value_name='Percentage')
plot_data['Day_Type'] = plot_data['Day_Type'].map({'%_No_Maint': 'Ngày thường', '%_With_Maint': 'Ngày có Bảo trì'})

plt.figure(figsize=(14, 8))
sns.barplot(x='Percentage', y='Downtime_Category', hue='Day_Type', data=plot_data, palette='muted')
plt.title('Biểu đồ 2: "Setup/Adjustment" là nguyên nhân chính gây Downtime sau Bảo trì', fontsize=16, fontweight='bold')
plt.xlabel('Tỷ trọng Downtime (%)')
plt.ylabel('Nhóm Nguyên nhân')
plt.legend(title='Loại ngày')
plt.grid(axis='x', linestyle='--')
plt.savefig(os.path.join(output_dir, "plot_maint_2_downtime_structure.png"), dpi=150, bbox_inches='tight')
plt.show()


# --- HỒI 3: KHOANH VÙNG "Ổ DỊCH" ---
def get_machine_family(line):
    line = str(line)
    if str(line).startswith('MAFT'): return 'MAFT'
    if str(line).startswith('MKBC'): return 'MKBC'
    if str(line).startswith('MKZA'): return 'MKZA'
    if str(line).startswith('MHPT'): return 'MHPT'
    if str(line).startswith('MKTU'): return 'MKTU'
    if str(line).startswith('MKFX'): return 'MKFX'
    return 'Other'
shift_summary['Machine_Family'] = shift_summary['LINE_NAME'].apply(get_machine_family)

maint_day_report = shift_summary[shift_summary['Is_Maint_Day']].groupby('Machine_Family').agg(
    Num_Maint_Shifts=('LINE_NAME', 'count'),
    Avg_Defect_Rate_On_Maint_Days=('Defect_Rate', 'mean')
).reset_index()
maint_day_report['Avg_Defect_Rate_On_Maint_Days_%'] = (maint_day_report['Avg_Defect_Rate_On_Maint_Days'] * 100)

print("="*80)
print("BẢNG 3: MỨC ĐỘ ẢNH HƯỞNG CỦA BẢO TRÌ LÊN TỪNG DÒNG MÁY")
print("="*80)
print(maint_day_report.sort_values('Avg_Defect_Rate_On_Maint_Days_%', ascending=False).to_string(index=False))
print("\n")

# BIỂU ĐỒ 3
plt.figure(figsize=(12, 7))
sns.barplot(x='Machine_Family', y='Avg_Defect_Rate_On_Maint_Days_%', data=maint_day_report.sort_values('Avg_Defect_Rate_On_Maint_Days_%', ascending=False), palette='Reds_r')
plt.title('Biểu đồ 3: MKBC và MKZA là hai dòng máy bị ảnh hưởng nặng nề nhất sau Bảo trì', fontsize=16)
plt.ylabel('Tỷ lệ lỗi trung bình trong ngày có Bảo trì (%)')
plt.xlabel('Dòng máy')
plt.savefig(os.path.join(output_dir, "plot_maint_3_impact_by_family.png"), dpi=150, bbox_inches='tight')
plt.show()

print(f"\n✅ Đã tạo và lưu 3 biểu đồ vào thư mục '{output_dir}'.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# === LOAD AND PREPARE DATA ===
df = pd.read_csv('../data/processed/master_data.csv', low_memory=False)
df['PRODUCTIONDATE_DAY_LOC'] = pd.to_datetime(df['PRODUCTIONDATE_DAY_LOC'])
df['Month'] = df['PRODUCTIONDATE_DAY_LOC'].dt.to_period('M')
df['Total_Production'] = df['GOOD_PRODUCTION_QTY'] + df['REJECT_PRODUCTION_QTY']
df['Has_Maintenance'] = df['ORDER_TYPE'] != 'No Maintenance Event'

# Focus only on MKZA production lines
mkza_df = df[df['LINE_NAME'].str.startswith('MKZA', na=False)].copy()

# === AGGREGATE MONTHLY METRICS FOR MKZA ===
mkza_timeline = mkza_df.groupby('Month').agg(
    # Count the number of shifts with maintenance activities
    Num_Maintenance_Shifts=('Has_Maintenance', lambda x: x.sum()),
    # Total downtime caused by Rework
    Total_Rework_Downtime=('DOWNTIME', lambda x: x[mkza_df.loc[x.index, 'UTIL_REASON_DESCRIPTION'].str.contains('Rework', na=False)].sum()),
    # Total downtime overall
    Total_Downtime=('DOWNTIME', 'sum'),
    # Quality indicators
    Total_Reject=('REJECT_PRODUCTION_QTY', 'sum'),
    Total_Production=('Total_Production', 'sum')
).reset_index()

# === DERIVED METRICS ===
mkza_timeline['Defect_Rate_%'] = (mkza_timeline['Total_Reject'] / mkza_timeline['Total_Production'] * 100).round(2)
mkza_timeline['Rework_%_of_Downtime'] = (mkza_timeline['Total_Rework_Downtime'] / mkza_timeline['Total_Downtime'] * 100).round(2)
mkza_timeline['Month_Str'] = mkza_timeline['Month'].astype(str)

# === DISPLAY THE MONTHLY TIMELINE TABLE ===
print("=== MKZA MONTHLY PERFORMANCE TIMELINE ===")
print("This table shows the relationship between Maintenance, Rework, and Quality over time.")
columns_to_show = [
    'Month_Str',
    'Num_Maintenance_Shifts',
    'Rework_%_of_Downtime',
    'Defect_Rate_%'
]
print(mkza_timeline[columns_to_show].to_string(index=False))
print("\n")

# === DUAL-AXIS CHART: VISUALIZING THE RELATIONSHIP ===
fig, ax1 = plt.subplots(figsize=(14, 8))

# Axis 1: Plot % Rework and Defect Rate
sns.lineplot(data=mkza_timeline, x='Month_Str', y='Rework_%_of_Downtime',
             color='red', marker='o', ax=ax1, label='% Downtime due to Rework')
sns.lineplot(data=mkza_timeline, x='Month_Str', y='Defect_Rate_%',
             color='orange', marker='s', ax=ax1, label='Defect Rate (%)')
ax1.set_xlabel('Month')
ax1.set_ylabel('Percentage (%)', color='black')
ax1.tick_params(axis='y', labelcolor='black')
ax1.legend(loc='upper left')

# Axis 2: Plot the number of Maintenance Shifts
ax2 = ax1.twinx()
sns.barplot(data=mkza_timeline, x='Month_Str', y='Num_Maintenance_Shifts',
            color='skyblue', alpha=0.5, ax=ax2, label='Maintenance Shifts')
ax2.set_ylabel('Number of Maintenance Shifts', color='blue')
ax2.tick_params(axis='y', labelcolor='blue')
ax2.legend(loc='upper right')

plt.title('Relationship between Maintenance Activities and Rework Improvement on MKZA Line',
          fontsize=16, fontweight='bold')
fig.tight_layout()
plt.grid(False)  # Disable Seaborn grid for a cleaner look
plt.savefig("mkza_timeline_analysis.png", dpi=150)
plt.show()
```

```python
import pandas as pd
import numpy as np

# === LOAD AND PREPARE DATA ===
df = pd.read_csv('../data/processed/master_data.csv', low_memory=False)
df['PRODUCTIONDATE_DAY_LOC'] = pd.to_datetime(df['PRODUCTIONDATE_DAY_LOC'])
df['Month'] = df['PRODUCTIONDATE_DAY_LOC'].dt.month

# Focus only on MKZA production lines
mkza_df = df[df['LINE_NAME'].str.startswith('MKZA', na=False)].copy()

# --- TASK 1: CONTENT OF "APRIL MAINTENANCE CAMPAIGN" ---

print("=== TASK 1: 'MAINTENANCE RECORDS' OF MKZA IN APRIL ===\n")

# Filter maintenance events on MKZA in April
maint_april_mkza = mkza_df[
    (mkza_df['Month'] == 4) &
    (mkza_df['ORDER_TYPE'] != 'No Maintenance Event')
][['LINE_NAME', 'ORDER_TYPE', 'DESCRIPTION']].drop_duplicates()

if maint_april_mkza.empty:
    print("No maintenance activities were recorded for MKZA in April.")
else:
    print("Table: Maintenance activities performed on MKZA during April:")
    display(maint_april_mkza.sort_values(by='LINE_NAME'))

# --- TASK 2: PERFORMANCE COMPARISON BEFORE AND AFTER THE "CAMPAIGN" ---

print("\n=== TASK 2: EVALUATING PERFORMANCE IMPROVEMENT ON EACH MKZA MACHINE ===\n")

# Define time periods
# "Before" period: January, February, March
# "After" period: April, May, June
mkza_df['Period'] = np.where(mkza_df['Month'].isin([1, 2, 3]), 'Before (Jan-Mar)', 'After (Apr-Jun)')

# Calculate Rework % by period for each machine
performance_comparison = mkza_df.groupby(['LINE_NAME', 'Period']).agg(
    Total_Downtime=('DOWNTIME', 'sum'),
    Rework_Downtime=('DOWNTIME', lambda x: x[mkza_df.loc[x.index, 'UTIL_REASON_DESCRIPTION'].str.contains('Rework', na=False)].sum())
).reset_index()

# Keep only machines that were active in both periods for a fair comparison
valid_lines = performance_comparison.groupby('LINE_NAME').filter(lambda x: len(x) == 2)['LINE_NAME'].unique()
performance_comparison = performance_comparison[performance_comparison['LINE_NAME'].isin(valid_lines)]

# Create pivot table for comparison
comparison_pivot = performance_comparison.pivot_table(
    index='LINE_NAME',
    columns='Period',
    values=['Total_Downtime', 'Rework_Downtime'],
    fill_value=0
)

# Compute % Rework for each period
comparison_pivot['Rework_%_Before'] = (
    comparison_pivot[('Rework_Downtime', 'Before (Jan-Mar)')] /
    comparison_pivot[('Total_Downtime', 'Before (Jan-Mar)')] * 100
).fillna(0)
comparison_pivot['Rework_%_After'] = (
    comparison_pivot[('Rework_Downtime', 'After (Apr-Jun)')] /
    comparison_pivot[('Total_Downtime', 'After (Apr-Jun)')] * 100
).fillna(0)

# Calculate improvement points
comparison_pivot['Improvement_(points)'] = (
    comparison_pivot['Rework_%_Before'] - comparison_pivot['Rework_%_After']
)

# Sort by improvement level
final_report = comparison_pivot[
    ['Rework_%_Before', 'Rework_%_After', 'Improvement_(points)']
].sort_values(by='Improvement_(points)', ascending=False)

print("Table: Comparison of Rework-Related Downtime Rates BEFORE and AFTER the April Maintenance Campaign (per MKZA machine)")
display(final_report.round(1))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# --- 0. SETUP & CREATE OUTPUT FOLDER ---
output_dir = "mkza_case_study"
if not os.path.exists(output_dir):
    os.makedirs(output_dir)
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 7)
plt.rcParams['font.size'] = 12

# --- 1. LOAD DATA & PREPARE FOR CASE STUDY ---
try:
    df = pd.read_csv('../data/processed/master_data.csv', low_memory=False)
    print("✅ Successfully loaded '../data/processed/master_data.csv'.")
except FileNotFoundError:
    print("❌ ERROR: File '../data/processed/master_data.csv' not found.")
    exit()

df['PRODUCTIONDATE_DAY_LOC'] = pd.to_datetime(df['PRODUCTIONDATE_DAY_LOC'])
df['Month'] = df['PRODUCTIONDATE_DAY_LOC'].dt.month
df['Total_Production'] = df['GOOD_PRODUCTION_QTY'] + df['REJECT_PRODUCTION_QTY']  # Ensure Total_Production exists
df['Has_Maintenance'] = df['ORDER_TYPE'] != 'No Maintenance Event'  # Create before filtering

# Filter only MKZA lines
mkza_df = df[df['LINE_NAME'].str.startswith('MKZA', na=False)].copy()
print("✅ Data filtered and prepared for MKZA Case Study.\n")

# --- 2. ACT I: DIAGNOSIS (Q1 DATA) ---
period1_df = mkza_df[mkza_df['Month'].isin([1, 2, 3])]
total_dt_p1 = period1_df['DOWNTIME'].sum()
rework_dt_p1 = period1_df[period1_df['UTIL_REASON_DESCRIPTION'].astype(str).str.contains('Rework', na=False)]['DOWNTIME'].sum()
rework_percent_p1 = (rework_dt_p1 / total_dt_p1 * 100) if total_dt_p1 > 0 else 0

print("="*80)
print("ACT I: DIAGNOSIS OF MKZA (Q1)")
print("="*80)
print(f"- Total Downtime (Q1): {total_dt_p1:,.0f} minutes")
print(f"- Downtime due to Rework (Q1): {rework_dt_p1:,.0f} minutes")
print(f"- Rework Share in Q1: {rework_percent_p1:.1f}%")
print("\n")

# --- 3. ACT II: “THE TREATMENT” (APRIL MAINTENANCE CAMPAIGN) ---
maint_april_mkza = mkza_df[
    (mkza_df['Month'] == 4) &
    (mkza_df['ORDER_TYPE'] != 'No Maintenance Event')
][['LINE_NAME', 'ORDER_TYPE', 'DESCRIPTION']].drop_duplicates().sort_values(by='LINE_NAME')

print("="*80)
print("ACT II: “THE TREATMENT” – APRIL MAINTENANCE CAMPAIGN")
print("="*80)
print("Table: Preventive Maintenance (PM10) activities performed:")
display(maint_april_mkza)
print("\n")

# --- 4. ACT III: EFFECTIVENESS EVALUATION (BEFORE & AFTER COMPARISON) ---
mkza_df['Period'] = np.where(mkza_df['Month'].isin([1, 2, 3]), 'Before (Q1)', 'After (Q2)')
performance_comparison = mkza_df.groupby(['LINE_NAME', 'Period']).agg(
    Total_Downtime=('DOWNTIME', 'sum'),
    Rework_Downtime=('DOWNTIME', lambda x: x[mkza_df.loc[x.index, 'UTIL_REASON_DESCRIPTION'].str.contains('Rework', na=False)].sum())
).reset_index()

# Keep only lines that have data in both periods
valid_lines = performance_comparison.groupby('LINE_NAME').filter(lambda x: len(x) == 2)['LINE_NAME'].unique()
performance_comparison = performance_comparison[performance_comparison['LINE_NAME'].isin(valid_lines)]

comparison_pivot = performance_comparison.pivot_table(index='LINE_NAME', columns='Period', values='Rework_Downtime', fill_value=0)
total_dt_pivot = performance_comparison.pivot_table(index='LINE_NAME', columns='Period', values='Total_Downtime', fill_value=0)

# Calculate Rework %
comparison_pivot['Rework_%_Before'] = (comparison_pivot['Before (Q1)'] / total_dt_pivot['Before (Q1)'] * 100).fillna(0)
comparison_pivot['Rework_%_After'] = (comparison_pivot['After (Q2)'] / total_dt_pivot['After (Q2)'] * 100).fillna(0)
comparison_pivot['Improvement (percentage points)'] = comparison_pivot['Rework_%_Before'] - comparison_pivot['Rework_%_After']

final_report = comparison_pivot[['Rework_%_Before', 'Rework_%_After', 'Improvement (percentage points)']].sort_values(by='Improvement (percentage points)', ascending=False)

print("="*80)
print("ACT III: EFFECTIVENESS EVALUATION – IMPROVEMENT BY MACHINE")
print("="*80)
print("Table: Comparison of Rework Downtime Ratio BEFORE vs AFTER the April campaign")
display(final_report.round(1))

# BAR CHART: BEFORE vs AFTER
plot_data = final_report[['Rework_%_Before', 'Rework_%_After']].reset_index()
plot_data_melted = plot_data.melt(id_vars='LINE_NAME', var_name='Period', value_name='Rework Ratio (%)')

plt.figure(figsize=(14, 8))
sns.barplot(x='Rework Ratio (%)', y='LINE_NAME', hue='Period', data=plot_data_melted, palette=['coral', 'skyblue'])
plt.title('Chart: Clear Effectiveness of Maintenance – Sharp Reduction in Rework Downtime across MKZA Lines', fontsize=16, fontweight='bold')
plt.xlabel('Rework Downtime Ratio (%)')
plt.ylabel('Line')
plt.legend(title='Period')
plt.grid(axis='x', linestyle='--')
plt.savefig(os.path.join(output_dir, "plot_mkza_before_after.png"), dpi=150, bbox_inches='tight')
plt.show()

# LINE CHART: OVERALL TREND
mkza_timeline = mkza_df.groupby('Month').agg(
    Num_Maintenance_Shifts=('Has_Maintenance', lambda x: x.sum()),  # Count True values
    Total_Rework_Downtime=('DOWNTIME', lambda x: x[mkza_df.loc[x.index, 'UTIL_REASON_DESCRIPTION'].str.contains('Rework', na=False)].sum()),
    Total_Downtime=('DOWNTIME', 'sum')
).reset_index()
mkza_timeline['Rework_%_of_Downtime'] = (mkza_timeline['Total_Rework_Downtime'] / mkza_timeline['Total_Downtime'] * 100).fillna(0)
mkza_timeline['Month_Str'] = mkza_timeline['Month'].astype(str)

fig, ax1 = plt.subplots(figsize=(14, 8))
sns.lineplot(data=mkza_timeline, x='Month_Str', y='Rework_%_of_Downtime', color='red', marker='o', ax=ax1, label='% Downtime due to Rework')
ax1.set_xlabel('Month')
ax1.set_ylabel('Rework Downtime Ratio (%)', color='red')
ax1.tick_params(axis='y', labelcolor='red')
ax1.legend(loc='upper left')
ax2 = ax1.twinx()
sns.barplot(data=mkza_timeline, x='Month_Str', y='Num_Maintenance_Shifts', color='skyblue', alpha=0.5, ax=ax2, label='Number of Maintenance Shifts')
ax2.set_ylabel('Number of Maintenance Shifts', color='blue')
ax2.tick_params(axis='y', labelcolor='blue')
ax2.legend(loc='upper right')
plt.title('Chart: Cause–Effect Relationship between Maintenance and Rework Improvement on MKZA', fontsize=16, fontweight='bold')
fig.tight_layout()
plt.savefig(os.path.join(output_dir, "plot_mkza_timeline_correlation.png"), dpi=150)
plt.show()

print(f"\n✅ All evidence for MKZA Case Study has been generated and saved to the '{output_dir}' folder.")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# === SETUP ===
sns.set_style("whitegrid")
plt.rcParams.update({'font.size': 10})

# === LOAD RAW DATA ===
print("Loading raw data...")
prod = pd.read_csv('../data/raw/production_logs.csv', low_memory=False)
maint = pd.read_csv('../data/raw/maintenance_order.csv')
cross = pd.read_csv('../data/raw/cross_reference.csv')

# === DATA PREPROCESSING ===
# Normalize date formats
prod['Date'] = pd.to_datetime(prod['PRODUCTIONDATE_DAY_LOC']).dt.date
maint['Date'] = pd.to_datetime(maint['BASIC_START_DATE']).dt.date
cross.rename(columns={'OPERA NAME': 'LINE_NAME'}, inplace=True)

# Attach LINE_NAME to maintenance records
maint = maint.merge(cross[['EQUIPMENT_ID', 'LINE_NAME']], on='EQUIPMENT_ID', how='left')
maint = maint[['Date', 'LINE_NAME']].dropna().drop_duplicates()
maint['Has_Maintenance'] = True

# Merge into production data
prod = prod.merge(maint, on=['Date', 'LINE_NAME'], how='left')
prod['Has_Maintenance'] = prod['Has_Maintenance'].fillna(False)

# Support columns
prod['Total_Production'] = prod['GOOD_PRODUCTION_QTY'] + prod['REJECT_PRODUCTION_QTY']
prod['Defect_Rate'] = np.where(
    prod['Total_Production'] > 0,
    prod['REJECT_PRODUCTION_QTY'] / prod['Total_Production'],
    np.nan
)
prod['Line_Prefix'] = prod['LINE_NAME'].str[:4]
prod['Is_Handle_Jam'] = prod['UTIL_REASON_DESCRIPTION'].astype(str).str.contains(
    r'Handle|Jammed handle|Handle stuck', case=False, na=False
)
prod['Has_Changeover'] = prod['CHANGEOVER_DURATION'] > 0

# === 1. TOP 10 DOWNTIME REASONS (Pareto) ===
top_downtime = prod.groupby('UTIL_REASON_DESCRIPTION')['DOWNTIME'].sum().sort_values(ascending=False).head(10)
plt.figure(figsize=(10, 6))
sns.barplot(x=top_downtime.values, y=top_downtime.index, palette='Reds_r')
plt.title('Top 10 Downtime Causes (Pareto)')
plt.xlabel('Total Downtime (minutes)')
plt.show()

# === 2. DEFECT RATE BY LINE ===
defect_by_line = prod.groupby('Line_Prefix').agg(
    Total_Good=('GOOD_PRODUCTION_QTY', 'sum'),
    Total_Reject=('REJECT_PRODUCTION_QTY', 'sum')
).reset_index()
defect_by_line['Defect_Rate'] = defect_by_line['Total_Reject'] / (
    defect_by_line['Total_Good'] + defect_by_line['Total_Reject']
)
plt.figure(figsize=(8, 5))
sns.barplot(
    data=defect_by_line.sort_values('Defect_Rate', ascending=False),
    x='Defect_Rate', y='Line_Prefix', palette='Blues_r'
)
plt.title('Defect Rate by Machine Line')
plt.xlabel('Defect Rate')
plt.show()

# === 3. HANDLE JAM DOWNTIME BY LINE ===
handle_downtime = prod[prod['Is_Handle_Jam']].groupby('Line_Prefix')['DOWNTIME'].sum().reset_index()
plt.figure(figsize=(8, 5))
sns.barplot(
    data=handle_downtime.sort_values('DOWNTIME', ascending=False),
    x='DOWNTIME', y='Line_Prefix', palette='Oranges_r'
)
plt.title('Downtime due to "Handle Jam" by Machine Line')
plt.xlabel('Total Downtime (minutes)')
plt.show()

# === 4. IMPACT OF CHANGEOVER ON HANDLE JAM (MKBC) ===
mkbc = prod[prod['Line_Prefix'] == 'MKBC'].copy()
mkbc['Shift_Key'] = mkbc['Date'].astype(str) + '_' + mkbc['SHIFT_NAME']
ca_summary = mkbc.groupby('Shift_Key').agg(
    Has_Changeover=('Has_Changeover', 'any'),
    Handle_Jam_Count=('Is_Handle_Jam', 'sum')
).reset_index()
avg_with = ca_summary[ca_summary['Has_Changeover']]['Handle_Jam_Count'].mean()
avg_without = ca_summary[~ca_summary['Has_Changeover']]['Handle_Jam_Count'].mean()
plt.figure(figsize=(6, 4))
plt.bar(['With Changeover', 'Without Changeover'], [avg_with, avg_without], color=['red', 'gray'])
plt.title('Average Handle Jam per Shift by Changeover (MKBC)')
plt.ylabel('Jam Count per Shift')
plt.show()

# === 5. DEFECT RATE: WITH VS WITHOUT MAINTENANCE ===
maint_effect = prod.groupby('Has_Maintenance')['Defect_Rate'].mean()
plt.figure(figsize=(6, 4))
plt.bar(['Without Maintenance', 'With Maintenance'], maint_effect.values, color=['green', 'red'])
plt.title('Defect Rate: With vs Without Maintenance')
plt.ylabel('Defect Rate')
plt.show()

# === 6. DOWNTIME BEFORE AND AFTER MAINTENANCE (3 days) ===
maint_dates = maint[['Date', 'LINE_NAME']].copy()
after_days = []
for i in range(1, 4):
    temp = maint_dates.copy()
    temp['Date'] = pd.to_datetime(temp['Date']) + pd.Timedelta(days=i)
    temp['Post_Maint'] = True
    after_days.append(temp)
after_df = pd.concat(after_days)
after_df['Date'] = after_df['Date'].dt.date

prod_check = prod[['Date', 'LINE_NAME', 'DOWNTIME']].copy()
prod_check = prod_check.merge(after_df, on=['Date', 'LINE_NAME'], how='left')
prod_check['Post_Maint'] = prod_check['Post_Maint'].fillna(False)

downtime_after = prod_check[prod_check['Post_Maint']]['DOWNTIME'].mean()
downtime_normal = prod_check[~prod_check['Post_Maint']]['DOWNTIME'].mean()
plt.figure(figsize=(6, 4))
plt.bar(['3 Days After Maintenance', 'Normal Days'], [downtime_after, downtime_normal], color=['orange', 'blue'])
plt.title('Downtime: After Maintenance vs Normal Days')
plt.ylabel('Average Downtime (minutes)')
plt.show()

# === 7. DEFECT RATE BY PRODUCT ON MAFT ===
maft = prod[prod['Line_Prefix'] == 'MAFT'].copy()
product_defect = maft.groupby('SIZE_TYPE').agg(
    Total_Good=('GOOD_PRODUCTION_QTY', 'sum'),
    Total_Reject=('REJECT_PRODUCTION_QTY', 'sum')
).reset_index()
product_defect['Defect_Rate'] = product_defect['Total_Reject'] / (
    product_defect['Total_Good'] + product_defect['Total_Reject']
)
plt.figure(figsize=(8, 4))
sns.barplot(
    data=product_defect.sort_values('Defect_Rate', ascending=False),
    x='Defect_Rate', y='SIZE_TYPE', palette='Purples_r'
)
plt.title('Defect Rate by Product on MAFT')
plt.xlabel('Defect Rate')
plt.show()

# === 8. HANDLE JAM TREND OVER TIME (MKBC) ===
mkbc_daily = mkbc.groupby('Date')['Is_Handle_Jam'].sum().reset_index()
plt.figure(figsize=(12, 4))
plt.plot(mkbc_daily['Date'], mkbc_daily['Is_Handle_Jam'], color='red', alpha=0.7)
plt.title('Trend of "Handle Jam" on MKBC over Time')
plt.ylabel('Jam Count per Day')
plt.show()

# === PRINT SUMMARY TABLES ===
print("="*80)
print("TABLE 1: TOP 10 DOWNTIME CAUSES")
print(top_downtime.to_string())
print("\n" + "="*80)
print("TABLE 2: DEFECT RATE BY MACHINE LINE")
print(defect_by_line[['Line_Prefix', 'Defect_Rate']].sort_values('Defect_Rate', ascending=False).to_string(index=False))
print("\n" + "="*80)
print("TABLE 3: HANDLE JAM DOWNTIME BY LINE")
print(handle_downtime.sort_values('DOWNTIME', ascending=False).to_string(index=False))
print("\n" + "="*80)
print("TABLE 4: IMPACT OF CHANGEOVER ON HANDLE JAM (MKBC)")
print(f"With Changeover: {avg_with:.3f} jams/shift")
print(f"Without Changeover: {avg_without:.3f} jams/shift")
print(f"Difference: {avg_with - avg_without:.3f} jams")
print("\n" + "="*80)
print("TABLE 5: MAINTENANCE EFFECT SUMMARY")
maint_summary = prod.groupby('Has_Maintenance').agg(
    Avg_Defect_Rate=('Defect_Rate', 'mean'),
    Avg_Downtime=('DOWNTIME', 'mean'),
    Count=('DOWNTIME', 'size')
).reset_index()
maint_summary['Has_Maintenance'] = maint_summary['Has_Maintenance'].map({True: 'With Maintenance', False: 'Without Maintenance'})
print(maint_summary.to_string(index=False))
print("\n" + "="*80)
print("TABLE 6: DEFECT RATE BY PRODUCT ON MAFT")
print(product_defect[['SIZE_TYPE', 'Defect_Rate']].sort_values('Defect_Rate', ascending=False).to_string(index=False))
print("\n" + "="*80)
print("TABLE 7: TOP 5 PRODUCTS CAUSING HANDLE JAM ON MKBC")
jam_by_product = mkbc[mkbc['Is_Handle_Jam']].groupby('SIZE_TYPE')['DOWNTIME'].sum().sort_values(ascending=False).head(5)
print(jam_by_product.to_string())
print("\n" + "="*80)
print("TABLE 8: MAINTENANCE STATISTICAL TEST")
with_m = prod[prod['Has_Maintenance']]
without_m = prod[~prod['Has_Maintenance']]
good1 = with_m['GOOD_PRODUCTION_QTY'].sum()
bad1 = with_m['REJECT_PRODUCTION_QTY'].sum()
good2 = without_m['GOOD_PRODUCTION_QTY'].sum()
bad2 = without_m['REJECT_PRODUCTION_QTY'].sum()
dr1 = bad1 / (good1 + bad1)
dr2 = bad2 / (good2 + bad2)
print(f"Defect rate (with maintenance): {dr1:.4f}")
print(f"Defect rate (without maintenance): {dr2:.4f}")
print(f"Relative difference: {(dr1/dr2 - 1)*100:.1f}%")
print("\n" + "="*80)
print("LIST OF PROCESSES REQUIRING IMPROVEMENT")
print("1. Changeover Process on MKBC")
print("2. Operation Process on MAFT Line")
print("3. Maintenance & Post-Maintenance Inspection")
print("4. Product Quality Control Process")
print("5. Raw Material Management Process")
print("\n✅ All 8 charts have been displayed successfully.")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# --- 0. SETUP & CREATE OUTPUT DIRECTORY ---
output_dir = "mkbc_deep_dive"
if not os.path.exists(output_dir):
    os.makedirs(output_dir)
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 7)
plt.rcParams['font.size'] = 12

# --- 1. LOAD DATA & PREPARE FOR MKBC ANALYSIS ---
try:
    df = pd.read_csv('../data/processed/master_data.csv', low_memory=False)
    print("✅ Successfully loaded '../data/processed/master_data.csv'.")
except FileNotFoundError:
    print("❌ ERROR: File '../data/processed/master_data.csv' not found.")
    exit()

# Filter only MKBC data
mkbc_df = df[df['LINE_NAME'].str.startswith('MKBC', na=False)].copy()
mkbc_df['Total_Production'] = mkbc_df['GOOD_PRODUCTION_QTY'] + mkbc_df['REJECT_PRODUCTION_QTY']

# Identify handle jam events
jam_keywords = ['jam', 'stuck', 'double', 'lack of']
mkbc_df['Is_Handle_Jam'] = mkbc_df['UTIL_REASON_DESCRIPTION'].str.contains('handle', case=False, na=False) & \
                           mkbc_df['UTIL_REASON_DESCRIPTION'].str.contains('|'.join(jam_keywords), case=False, na=False)
mkbc_df['Handle_Jam_Downtime'] = mkbc_df['DOWNTIME'].where(mkbc_df['Is_Handle_Jam'], 0)
print("✅ Filtered and prepared data for MKBC case study.\n")


# --- 2. ACT I: DIAGNOSING THE 'ILLNESS' OF MKBC ---

# TABLE 1: General “Health Check” of MKBC
total_dt_mkbc = mkbc_df['DOWNTIME'].sum()
jam_dt_mkbc = mkbc_df['Handle_Jam_Downtime'].sum()
jam_percent = (jam_dt_mkbc / total_dt_mkbc * 100) if total_dt_mkbc > 0 else 0

total_reject_mkbc = mkbc_df['REJECT_PRODUCTION_QTY'].sum()
total_prod_mkbc = mkbc_df['Total_Production'].sum()
defect_rate_mkbc = (total_reject_mkbc / total_prod_mkbc * 100) if total_prod_mkbc > 0 else 0

table1_data = {
    'Indicator': ['Total Downtime (min)', 'Downtime due to Handle Jam (min)', '% Downtime from Handle Jam', 'Overall Defect Rate (%)'],
    'Value': [f"{total_dt_mkbc:,.0f}", f"{jam_dt_mkbc:,.0f}", f"{jam_percent:.1f}%", f"{defect_rate_mkbc:.2f}%"]
}
table1_df = pd.DataFrame(table1_data)

print("="*80)
print("ACT I: DIAGNOSING THE 'ILLNESS' OF MKBC")
print("="*80)
print(table1_df.to_string(index=False))
print("\n")


# CHART 1: Pie chart of MKBC Downtime distribution
other_dt_mkbc = total_dt_mkbc - jam_dt_mkbc
plt.figure(figsize=(8, 8))
plt.pie([jam_dt_mkbc, other_dt_mkbc],
        labels=['Downtime due to Handle Jam', 'Other Downtime'],
        autopct='%1.1f%%', startangle=90, colors=['red', 'skyblue'],
        wedgeprops={'edgecolor':'black'}, textprops={'fontweight':'bold', 'fontsize':14})
plt.title('Chart 1: Nearly half of MKBC Downtime comes from "Handle Jam"', fontsize=16, fontweight='bold')
plt.savefig(os.path.join(output_dir, "plot_mkbc_1_downtime_pie.png"), dpi=150, bbox_inches='tight')
plt.show()


# --- 3. ACT II: “DISSECTING” ROOT CAUSES ---

# TABLE 2: Top 10 detailed “Handle Jam” error codes
table2_df = mkbc_df[mkbc_df['Is_Handle_Jam']].groupby('UTIL_REASON_DESCRIPTION')['DOWNTIME'] \
    .sum().sort_values(ascending=False).head(10).to_frame('Total Downtime (min)')
print("="*80)
print("ACT II: 'DISSECTING' HANDLE JAM ROOT CAUSES")
print("="*80)
print("Table 2: Top 10 detailed error descriptions related to 'Handle Jam'")
display(table2_df)


# CHART 2: Pareto chart for detailed error codes
plt.figure(figsize=(12, 8))
sns.barplot(y=table2_df.index, x=table2_df['Total Downtime (min)'], palette='Reds_r')
plt.title('Chart 2: “Handle Jam” issues mainly occur at End Feeder and FM Chain stages', fontsize=16)
plt.xlabel('Total Downtime (min)')
plt.ylabel('Detailed Error Description')
plt.savefig(os.path.join(output_dir, "plot_mkbc_2_jam_details_pareto.png"), dpi=150, bbox_inches='tight')
plt.show()

# TABLE 3: Top 10 MKBC machines with highest Handle Jam downtime
table3_df = mkbc_df.groupby('LINE_NAME')['Handle_Jam_Downtime'].sum().sort_values(ascending=False).head(10).to_frame('Total Jam Downtime (min)')
print("\nTable 3: Top 10 MKBC “patients” most affected by Handle Jam issues")
display(table3_df)


# --- 4. ACT III: IDENTIFYING THE 'CULPRITS' ---

# TABLE 4: Top 10 products causing the most Handle Jam on MKBC
table4_df = mkbc_df[mkbc_df['Is_Handle_Jam']].groupby('SIZE_TYPE')['DOWNTIME'] \
    .sum().sort_values(ascending=False).head(10).to_frame('Total Jam Downtime (min)')
print("="*80)
print("ACT III: IDENTIFYING THE 'CULPRITS'")
print("="*80)
print("Table 4: Top 10 products causing the most Handle Jam on MKBC")
display(table4_df)


# CHART 3: Bar chart showing high-risk products
plt.figure(figsize=(12, 8))
sns.barplot(y=table4_df.index, x=table4_df['Total Jam Downtime (min)'], palette='Oranges_r')
plt.title('Chart 3: “Handle Jam” issues are more severe for certain products', fontsize=16)
plt.xlabel('Total Downtime due to Handle Jam (min)')
plt.ylabel('Product Name')
plt.savefig(os.path.join(output_dir, "plot_mkbc_3_jam_by_product.png"), dpi=150, bbox_inches='tight')
plt.show()

# TABLE 5: Performance comparison between “High-Risk” and other products
risky_products = table4_df.head(5).index.tolist()
prod_events_mkbc = prod_events[prod_events['LINE_NAME'].str.startswith('MKBC', na=False)]
prod_events_mkbc['Risk_Group'] = np.where(prod_events_mkbc['SIZE_TYPE'].isin(risky_products),
                                          'High-Risk Products', 'Other Products')

table5_df = prod_events_mkbc.groupby('Risk_Group').agg(
    Avg_Defect_Rate=('Defect_Rate', 'mean'),
    Total_Production=('Total_Production', 'sum')
)
table5_df['Avg_Defect_Rate_%'] = (table5_df['Avg_Defect_Rate'] * 100)
print("\nTable 5: Comparison of Defect Rates between 'High-Risk' and Other Product Groups")
display(table5_df)

print(f"\n✅ Evidence and visualizations for MKBC Case Study have been saved to folder '{output_dir}'.")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import re

# --- 0. SETUP & CREATE OUTPUT DIRECTORY ---
output_dir = "mkbc_deep_dive_v2"
if not os.path.exists(output_dir):
    os.makedirs(output_dir)
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 7)
plt.rcParams['font.size'] = 12

# --- 1. LOAD DATA & APPLY NEW CLASSIFICATION FUNCTION ---
try:
    df = pd.read_csv('../data/processed/master_data.csv', low_memory=False)
    print("✅ Successfully loaded '../data/processed/master_data.csv'.")
except FileNotFoundError:
    print("❌ ERROR: '../data/processed/master_data.csv' not found.")
    exit()

# NEW CLASSIFICATION FUNCTION
def classify_downtime_cause_by_code(reason):
    """Classify downtime causes based on standardized error codes."""
    if pd.isna(reason) or not str(reason).strip():
        return 'Unknown / Not Classified'

    text = str(reason).strip().lower()
    match = re.match(r'^\s*(\d{1,2})\s*-', text)
    code = int(match.group(1)) if match else None

    if code is None:
        # Handle non-coded descriptions
        if 'rework' in text:
            return 'Process & Setup / Changeover'  # Rework belongs to process category
        return 'Unknown / Not Classified'

    if code in [6, 7, 9, 10, 11, 12, 13, 14, 16, 17, 18, 19, 20, 22, 23, 25, 26, 27]:
        return 'Equipment / Mechanical Failure'
    if code in [2, 3, 8]:
        return 'Process & Setup / Changeover'
    if code in [4, 5, 24, 28]:
        return 'Supporting Systems Failure'
    if code in [15, 21, 29]:
        return 'Electrical / Control / Sensor Failure'
    if code in [1]:
        return 'Safety Stops'

    return 'Unknown / Not Classified'

df['Downtime_Category_Standard'] = df['UTIL_REASON_DESCRIPTION'].apply(classify_downtime_cause_by_code)

# Filter MKBC data only
mkbc_df = df[df['LINE_NAME'].str.startswith('MKBC', na=False)].copy()
mkbc_df['Total_Production'] = mkbc_df['GOOD_PRODUCTION_QTY'] + mkbc_df['REJECT_PRODUCTION_QTY']
print("✅ Filtered and prepared data for MKBC Case Study (v2).\n")


# --- 2. ACT I: DIAGNOSING THE “ILLNESS” OF MKBC (WITH STANDARDIZED CLASSIFICATION) ---

# TABLE 1: Overview of MKBC Downtime by Standardized Cause Category
table1_df = (
    mkbc_df[mkbc_df['DOWNTIME'] > 0]
    .groupby('Downtime_Category_Standard')['DOWNTIME']
    .sum()
    .sort_values(ascending=False)
    .to_frame('Total Downtime (min)')
)
table1_df['Percentage'] = (
    table1_df['Total Downtime (min)'] / table1_df['Total Downtime (min)'].sum() * 100
).map('{:.1f}%'.format)

print("="*80)
print("ACT I: DIAGNOSING THE 'ILLNESS' OF MKBC (WITH STANDARDIZED CLASSIFICATION)")
print("="*80)
print("Table 1: Distribution of MKBC Downtime by Standard Cause Categories")
display(table1_df)
print("\n")


# CHART 1: Pie chart showing MKBC Downtime distribution (Standardized categories)
plt.figure(figsize=(10, 8))
plt.pie(
    table1_df['Total Downtime (min)'],
    labels=table1_df.index,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops={'edgecolor': 'black'},
    textprops={'fontweight': 'bold', 'fontsize': 12},
    pctdistance=0.85
)
plt.title('Chart 1: Nearly 80% of MKBC Downtime comes from Equipment & Mechanical Failures', fontsize=16, fontweight='bold')
centre_circle = plt.Circle((0, 0), 0.70, fc='white')
fig = plt.gcf()
fig.gca().add_artist(centre_circle)
plt.savefig(os.path.join(output_dir, "plot_mkbc_v2_1_downtime_pie.png"), dpi=150, bbox_inches='tight')
plt.show()


# --- 3. ACT II: “DISSECTING” THE EQUIPMENT / MECHANICAL FAILURE GROUP ---

# TABLE 2: Top 10 most frequent detailed error codes in the Mechanical Failure category
mechanical_df_mkbc = mkbc_df[mkbc_df['Downtime_Category_Standard'] == 'Equipment / Mechanical Failure']
table2_df = (
    mechanical_df_mkbc.groupby('UTIL_REASON_DESCRIPTION')['DOWNTIME']
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .to_frame('Total Downtime (min)')
)

print("="*80)
print("ACT II: 'DISSECTING' THE EQUIPMENT / MECHANICAL FAILURE GROUP")
print("="*80)
print("Table 2: Top 10 Detailed Mechanical Error Codes on MKBC")
display(table2_df)


# CHART 2: Pareto chart of detailed mechanical failure codes
plt.figure(figsize=(12, 8))
sns.barplot(y=table2_df.index, x=table2_df['Total Downtime (min)'], palette='Reds_r')
plt.title('Chart 2: Mechanical issues are concentrated at End Feeder and FM Chain stages', fontsize=16)
plt.xlabel('Total Downtime (min)')
plt.ylabel('Detailed Error Description')
plt.savefig(os.path.join(output_dir, "plot_mkbc_v2_2_mechanical_details.png"), dpi=150, bbox_inches='tight')
plt.show()


# TABLE 3: Top 10 MKBC machines most affected by Mechanical Failures
table3_df = (
    mechanical_df_mkbc.groupby('LINE_NAME')['DOWNTIME']
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .to_frame('Total Downtime from Mechanical Failures (min)')
)
print("\nTable 3: Top 10 MKBC 'patients' most affected by Mechanical Failures")
display(table3_df)


# --- 4. ACT III: IDENTIFYING RELATED “CULPRITS” (PRODUCTS) ---

# TABLE 4: Top 10 products causing the most Mechanical Failures on MKBC
table4_df = (
    mechanical_df_mkbc.groupby('SIZE_TYPE')['DOWNTIME']
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .to_frame('Downtime from Mechanical Failures (min)')
)

print("="*80)
print("ACT III: IDENTIFYING THE RELATED 'CULPRITS'")
print("="*80)
print("Table 4: Top 10 Products Associated with Mechanical Failures on MKBC")
display(table4_df)


# CHART 3: Bar chart showing “high-risk” products
plt.figure(figsize=(12, 8))
sns.barplot(y=table4_df.index, x=table4_df['Downtime from Mechanical Failures (min)'], palette='Oranges_r')
plt.title('Chart 3: Mechanical Failures are more severe when producing certain products', fontsize=16)
plt.xlabel('Total Downtime from Mechanical Failures (min)')
plt.ylabel('Product Name')
plt.savefig(os.path.join(output_dir, "plot_mkbc_v2_3_mechanical_by_product.png"), dpi=150, bbox_inches='tight')
plt.show()

print(f"\n✅ Evidence and visualizations for MKBC Case Study (Version 2) have been saved to folder '{output_dir}'.")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import linregress

# --- 0. SETUP & LOAD DATA ---
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 7)
plt.rcParams['font.size'] = 12

try:
    df = pd.read_csv('../data/processed/master_data.csv', low_memory=False)
    print("✅ Successfully loaded '../data/processed/master_data.csv'.")
except FileNotFoundError:
    print("❌ ERROR: File '../data/processed/master_data.csv' not found.")
    exit()

# Prepare necessary columns
df['PRODUCTIONDATE_DAY_LOC'] = pd.to_datetime(df['PRODUCTIONDATE_DAY_LOC'])
df['Total_Production'] = df['GOOD_PRODUCTION_QTY'] + df['REJECT_PRODUCTION_QTY']
df['Month_Period'] = df['PRODUCTIONDATE_DAY_LOC'].dt.to_period('M')

# Keep only production events
prod_events = df[df['Total_Production'] > 0].copy()

# --- 1. CALCULATE TRENDS FOR EACH PRODUCT ---
print("\nScanning the entire product catalog to identify rising defect rate trends...")

# Only consider products with total production > 500,000 units for reliability
significant_products = prod_events.groupby('SIZE_TYPE')['Total_Production'].sum()
target_products = significant_products[significant_products > 500000].index.tolist()

emerging_issues_list = []

for product_name in target_products:
    # Filter data for each product
    product_df = prod_events[prod_events['SIZE_TYPE'] == product_name]

    # Calculate monthly defect rates
    monthly_stats = product_df.groupby('Month_Period').agg(
        Total_Reject=('REJECT_PRODUCTION_QTY', 'sum'),
        Total_Production=('Total_Production', 'sum')
    ).reset_index()
    monthly_stats['Defect_Rate'] = monthly_stats['Total_Reject'] / monthly_stats['Total_Production']

    # Only compute trend if at least 4 months of data are available
    if len(monthly_stats) >= 4:
        monthly_stats = monthly_stats.sort_values('Month_Period')

        # Perform linear regression to find the slope of the trend
        # x = integer sequence representing time (0, 1, 2, ...)
        x_time = np.arange(len(monthly_stats))
        y_defect = monthly_stats['Defect_Rate'].values

        slope, intercept, r_value, p_value, std_err = linregress(x_time, y_defect)

        # --- CONDITIONS TO FLAG AN “EMERGING ISSUE” ---
        # 1. Positive slope (increasing trend)
        # 2. Slope magnitude > 0.001 (i.e., > 0.1% per month increase)
        # 3. Final defect rate must be higher than the initial defect rate

        is_emerging = (
            slope > 0.001 and
            monthly_stats['Defect_Rate'].iloc[-1] > monthly_stats['Defect_Rate'].iloc[0]
        )

        if is_emerging:
            emerging_issues_list.append({
                'Product': product_name,
                'Trend_Slope_per_Month': slope,
                'Start_Rate': monthly_stats['Defect_Rate'].iloc[0],
                'End_Rate': monthly_stats['Defect_Rate'].iloc[-1],
                'P_Value': p_value  # Statistical significance of the trend
            })

# --- 2. DISPLAY RESULTS ---
print("\n" + "="*80)
print("RESULTS: EMERGING QUALITY ISSUES DETECTION (RISING DEFECT RATE TRENDS)")
print("="*80)

if not emerging_issues_list:
    print("\n✅ No products were found with a consistent and significant upward defect rate trend.")
else:
    emerging_df = pd.DataFrame(emerging_issues_list)
    emerging_df = emerging_df.sort_values(by='Trend_Slope_per_Month', ascending=False)

    # Format table for readability
    emerging_df['Start_Rate_%'] = emerging_df['Start_Rate'].map('{:.2%}'.format)
    emerging_df['End_Rate_%'] = emerging_df['End_Rate'].map('{:.2%}'.format)
    emerging_df['P_Value'] = emerging_df['P_Value'].map('{:.3f}'.format)

    print("\n--- TABLE: PRODUCTS SHOWING INCREASING DEFECT RATE TRENDS ---")
    display(emerging_df[['Product', 'Start_Rate_%', 'End_Rate_%', 'Trend_Slope_per_Month', 'P_Value']])

    # --- 3. PLOT VISUAL ILLUSTRATION ---
    print("\n--- CHART: VISUALIZING EMERGING QUALITY ISSUE TRENDS ---")

    # Get monthly trend data for identified products
    plot_df = prod_events[prod_events['SIZE_TYPE'].isin(emerging_df['Product'])].copy()
    monthly_plot_df = plot_df.groupby(['Month_Period', 'SIZE_TYPE']).agg(
        Total_Reject=('REJECT_PRODUCTION_QTY', 'sum'),
        Total_Production=('Total_Production', 'sum')
    ).reset_index()
    monthly_plot_df['Defect_Rate_%'] = (
        monthly_plot_df['Total_Reject'] / monthly_plot_df['Total_Production'] * 100
    )
    monthly_plot_df['Month_Str'] = monthly_plot_df['Month_Period'].astype(str)

    plt.figure(figsize=(14, 8))
    sns.lineplot(
        data=monthly_plot_df,
        x='Month_Str',
        y='Defect_Rate_%',
        hue='SIZE_TYPE',
        marker='o',
        linewidth=2.5
    )

    # Add overall trendline for each product
    for product_name in emerging_df['Product']:
        subset = monthly_plot_df[monthly_plot_df['SIZE_TYPE'] == product_name]
        sns.regplot(
            x=np.arange(len(subset)),
            y=subset['Defect_Rate_%'],
            scatter=False,
            ci=None,
            label=f'Trend: {product_name}'
        )

    plt.title('Chart: Upward Defect Rate Trends Detected in “Emerging Issue” Products', fontsize=16)
    plt.xlabel('Month')
    plt.ylabel('Defect Rate (%)')
    plt.xticks(rotation=45, ha='right')
    plt.legend(title='Product')
    plt.grid(True, which='both', linestyle='--')
    plt.tight_layout()
    plt.savefig("plot_emerging_quality_issues.png", dpi=150, bbox_inches='tight')
    plt.show()
```
